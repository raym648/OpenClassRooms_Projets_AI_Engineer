```python

# ============================================================
# CELLULE 2 - IMPORT DES LIBRAIRIES ET CONFIGURATION
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

from pathlib import Path

# Configuration graphique
plt.style.use("default")
sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 100)

print("✅ Librairies importées avec succès")
```

```python

# ============================================================
# CELLULE 3 - VERIFICATION & UPLOAD DES DONNEES SOURCES
# ============================================================

from google.colab import files
from pathlib import Path

# Dossier des données brutes
RAW_DATA_PATH = Path(f"{PROJECT_ROOT}/data/raw")
RAW_DATA_PATH.mkdir(parents=True, exist_ok=True)

# Liste des fichiers attendus (dataset Home Credit)
expected_files = [
    "application_train.csv",
    "application_test.csv",
    "bureau.csv",
    "bureau_balance.csv",
    "credit_card_balance.csv",
    "HomeCredit_columns_description.csv",
    "installments_payments.csv",
    "POS_CASH_balance.csv",
    "previous_application.csv",
    "sample_submission.csv"
]

# Vérification des fichiers manquants
missing_files = [f for f in expected_files if not (RAW_DATA_PATH / f).exists()]

if missing_files:
    print("⚠️ Fichiers manquants détectés :", missing_files)
    print("📥 Veuillez uploader les fichiers manquants via la fenêtre Colab.")

    uploaded = files.upload()

    for name, data in uploaded.items():
        with open(RAW_DATA_PATH / name, "wb") as f:
            f.write(data)

    print("✅ Upload terminé")

    # ============================================================
    # ✅ Amélioration 1 – Vérification finale des fichiers
    # ============================================================
    still_missing = [f for f in expected_files if not (RAW_DATA_PATH / f).exists()]
    if still_missing:
        raise FileNotFoundError(
            f"❌ Fichiers toujours manquants après upload : {still_missing}"
        )
else:
    print("✅ Tous les fichiers sont présents")

# ============================================================
# ✅ Amélioration 2 – Message pédagogique MLOps
# ============================================================
print(
    "ℹ️ Vérification terminée : la présence des fichiers est confirmée.\n"
    "ℹ️ L'encodage et la lisibilité des CSV seront validés lors de l'étape de chargement des données."
)
```

```python

# ============================================================
# CELLULE 4 - CHARGEMENT & INSPECTION DES DATASETS (ROBUSTE)
# ============================================================

import pandas as pd

datasets = {}

for file in expected_files:
    if not file.endswith(".csv"):
        continue

    df_name = file.replace(".csv", "")
    file_path = RAW_DATA_PATH / file

    # Chargement robuste avec fallback d'encodage
    try:
        df = pd.read_csv(file_path, encoding="utf-8")
        encoding_used = "utf-8"
    except UnicodeDecodeError:
        df = pd.read_csv(file_path, encoding="latin1")
        encoding_used = "latin1"

    datasets[df_name] = df

    print(f"\n📂 {df_name} | encoding={encoding_used}")
    print(df.info())
    display(df.head())
```

```python

# ============================================================
# 🧪 CELLULE 5.1 – Analyse des valeurs manquantes : application_train
# ============================================================


df_name = "application_train"
df = datasets[df_name]

plt.figure(figsize=(14, 6))
msno.matrix(df)
plt.title(f"Visualisation des valeurs manquantes - {df_name}")
plt.savefig(f"{PROJECT_ROOT}/reports/e01_missing_values_{df_name}.png", bbox_inches="tight")
plt.show()

# Statistiques simples
missing_ratio = (df.isna().sum() / len(df)).sort_values(ascending=False)
display(missing_ratio[missing_ratio > 0].head(10))
```

```python

# ============================================================
# CELLULE 7.2 - NETTOYAGE & IMPUTATION PAR TABLE
# (Suppression manuelle → Sauvegarde)
# ============================================================


# ============================================================
# ÉTAPE 2 — PARAMÈTRES MANUELS DE SUPPRESSION / NETTOYAGE
# ============================================================

MANUAL_CLEANING_RULES = {

    "application_train": {
        "drop_columns": [
            # Exemple :
            # "FLAG_MOBIL",
        ],
        "drop_rows_conditions": [
            # Exemple :
            # ("DAYS_EMPLOYED", 365243),
        ],
        "exclude_from_imputation": [
            "TARGET"
        ]
    },

    "bureau": {
        "drop_columns": [],
        "drop_rows_conditions": [],
        "exclude_from_imputation": []
    },

    "installments_payments": {
        "drop_columns": [],
        "drop_rows_conditions": [
            # Exemple :
            # ("AMT_PAYMENT", 0),
        ],
        "exclude_from_imputation": []
    },

    # Par défaut pour les autres tables
}
```

```python

# ============================================================
# CELLULE 7.3 - NETTOYAGE & IMPUTATION PAR TABLE
# (Suppression manuelle → Sauvegarde)
# ============================================================


# ============================================================
# ÉTAPE 3 — SUPPRESSION / CORRECTION MANUELLE & IMPUTATION
# ============================================================


from pathlib import Path
import numpy as np
import pandas as pd

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed"
)

def clean_table_with_manual_rules(df, table_name, rules):
    df_clean = df.copy()

    print(f"\n🧹 Nettoyage manuel — {table_name}")

    # --------------------------------------------------------
    # 1. Suppression manuelle de colonnes
    # --------------------------------------------------------
    cols_to_drop = rules.get("drop_columns", [])
    if cols_to_drop:
        print(f"🗑️ Colonnes supprimées : {cols_to_drop}")
        df_clean = df_clean.drop(columns=cols_to_drop, errors="ignore")

    # --------------------------------------------------------
    # 2. Suppression manuelle de lignes (conditions)
    # --------------------------------------------------------
    conditions = rules.get("drop_rows_conditions", [])
    for col, value in conditions:
        if col in df_clean.columns:
            n_before = len(df_clean)
            df_clean = df_clean[df_clean[col] != value]
            print(f"🗑️ Lignes supprimées où {col} == {value} ({n_before - len(df_clean)})")

    # --------------------------------------------------------
    # 3. Suppression des doublons (contrôlée)
    # --------------------------------------------------------
    n_dup = df_clean.duplicated().sum()
    if n_dup > 0:
        print(f"🗑️ Doublons supprimés : {n_dup}")
        df_clean = df_clean.drop_duplicates()

    # --------------------------------------------------------
    # 4. Imputation raisonnée (sauf exclusions)
    # --------------------------------------------------------
    exclude_cols = rules.get("exclude_from_imputation", [])

    for col in df_clean.columns:
        if col in exclude_cols:
            continue

        if df_clean[col].isnull().sum() > 0:
            if df_clean[col].dtype in ["int64", "float64"]:
                df_clean[col] = df_clean[col].fillna(df_clean[col].median())
            else:
                df_clean[col] = df_clean[col].fillna("Unknown")

    return df_clean
```

```python

# ============================================================
# CELLULE 7.4 - NETTOYAGE & IMPUTATION PAR TABLE
# (Sauvegarde)
# ============================================================

import gc

cleaned_datasets = {}

for table_name, df in datasets.items():
    print("\n" + "=" * 80)
    print(f"📊 TABLE : {table_name}")
    print("=" * 80)

    rules = MANUAL_CLEANING_RULES.get(table_name, {})
    df_clean = clean_table_with_manual_rules(df, table_name, rules)

    # Vérification finale
    remaining_na = df_clean.isnull().sum().sum()
    print(f"🔎 Valeurs manquantes restantes : {remaining_na}")

    # Sauvegarde
    output_path = PROCESSED_DIR / f"e01_{table_name}_clean.csv"
    df_clean.to_csv(output_path, index=False)
    print(f"💾 Table sauvegardée : {output_path.name}")

    cleaned_datasets[table_name] = df_clean
    application_train_cleaned = cleaned_datasets["application_train"]

print("\n✅ Nettoyage manuel terminé pour toutes les tables")

gc.collect()
```

```python

# ============================================================
# CELLULE 8.1 - PRÉPARATION DES CLÉS & JOINTURE DES DATAFRAMES
# ============================================================

import pandas as pd
from pathlib import Path
import gc

gc.collect()

# ------------------------------------------------------------
# Chargement des tables nettoyées
# ------------------------------------------------------------

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed"
)

application_train = pd.read_csv(PROCESSED_DIR / "e01_application_train_clean.csv")
bureau = pd.read_csv(PROCESSED_DIR / "e01_bureau_clean.csv")
bureau_balance = pd.read_csv(PROCESSED_DIR / "e01_bureau_balance_clean.csv")
previous_application = pd.read_csv(PROCESSED_DIR / "e01_previous_application_clean.csv")
installments = pd.read_csv(PROCESSED_DIR / "e01_installments_payments_clean.csv")
pos_cash = pd.read_csv(PROCESSED_DIR / "e01_POS_CASH_balance_clean.csv")
credit_card = pd.read_csv(PROCESSED_DIR / "e01_credit_card_balance_clean.csv")

print("✅ Tables nettoyées chargées")
```

```python

# ============================================================
# CELLULE 8.2 - Agrégation du bureau_balance par SK_ID_BUREAU
# ============================================================

bureau_balance_agg = (
    bureau_balance
    .groupby("SK_ID_BUREAU")
    .agg({
        "MONTHS_BALANCE": ["min", "max", "count"]
    })
)

bureau_balance_agg.columns = [
    "BUREAU_BALANCE_MONTH_MIN",
    "BUREAU_BALANCE_MONTH_MAX",
    "BUREAU_BALANCE_COUNT"
]

bureau_balance_agg.reset_index(inplace=True)

# Jointure avec bureau
bureau = bureau.merge(
    bureau_balance_agg,
    on="SK_ID_BUREAU",
    how="left"
)

print("✅ bureau_balance agrégé et joint à bureau")
```

```python

# ============================================================
# CELLULE 8.3 - Agrégation bureau → niveau client (SK_ID_CURR)
# ============================================================

bureau_agg = (
    bureau
    .groupby("SK_ID_CURR")
    .agg({
        "DAYS_CREDIT": ["mean", "min"],
        "AMT_CREDIT_SUM": ["mean", "sum"],
        "CREDIT_DAY_OVERDUE": ["max"],
        "BUREAU_BALANCE_COUNT": ["sum"]
    })
)

bureau_agg.columns = [
    "BUREAU_DAYS_CREDIT_MEAN",
    "BUREAU_DAYS_CREDIT_MIN",
    "BUREAU_AMT_CREDIT_MEAN",
    "BUREAU_AMT_CREDIT_SUM",
    "BUREAU_MAX_OVERDUE",
    "BUREAU_BALANCE_LINES"
]

bureau_agg.reset_index(inplace=True)

print("✅ bureau agrégé au niveau client")
```

```python

# ============================================================
# CELLULE 8.4 - Agrégation des autres tables satellites : (previous_application)
# ============================================================

prev_agg = (
    previous_application
    .groupby("SK_ID_CURR")
    .agg({
        "AMT_APPLICATION": ["mean"],
        "AMT_CREDIT": ["mean"],
        "DAYS_DECISION": ["min"],
        "CNT_PAYMENT": ["mean"]
    })
)

prev_agg.columns = [
    "PREV_AMT_APPLICATION_MEAN",
    "PREV_AMT_CREDIT_MEAN",
    "PREV_DAYS_DECISION_MIN",
    "PREV_CNT_PAYMENT_MEAN"
]

prev_agg.reset_index(inplace=True)
```

```python

# ============================================================
# CELLULE 8.5 - Agrégation des autres tables satellites : (installments_payments)
# ============================================================

installments["PAYMENT_DIFF"] = (
    installments["AMT_PAYMENT"] - installments["AMT_INSTALMENT"]
)

inst_agg = (
    installments
    .groupby("SK_ID_CURR")
    .agg({
        "PAYMENT_DIFF": ["mean"],
        "DAYS_ENTRY_PAYMENT": ["mean"]
    })
)

inst_agg.columns = [
    "INST_PAYMENT_DIFF_MEAN",
    "INST_DAYS_ENTRY_MEAN"
]

inst_agg.reset_index(inplace=True)
```

```python

# ============================================================
# CELLULE 8.6 - Agrégation des autres tables satellites : (POS_CASH_balance)
# ============================================================

pos_agg = (
    pos_cash
    .groupby("SK_ID_CURR")
    .agg({
        "SK_DPD": ["mean", "max"],
        "MONTHS_BALANCE": ["count"]
    })
)

pos_agg.columns = [
    "POS_DPD_MEAN",
    "POS_DPD_MAX",
    "POS_LOAN_COUNT"
]

pos_agg.reset_index(inplace=True)
```

```python

# ============================================================
# CELLULE 8.7 - Agrégation des autres tables satellites : (credit_card_balance)
# ============================================================

cc_agg = (
    credit_card
    .groupby("SK_ID_CURR")
    .agg({
        "AMT_BALANCE": ["mean"],
        "AMT_CREDIT_LIMIT_ACTUAL": ["mean"],
        "SK_DPD": ["max"]
    })
)

cc_agg.columns = [
    "CC_BALANCE_MEAN",
    "CC_CREDIT_LIMIT_MEAN",
    "CC_MAX_DPD"
]

cc_agg.reset_index(inplace=True)
```

```python

# ============================================================
# CELLULE 8.8 - Jointure finale autour de application_train
# ============================================================

df_final = application_train.copy()

for agg_df in [
    bureau_agg,
    prev_agg,
    inst_agg,
    pos_agg,
    cc_agg
]:
    df_final = df_final.merge(
        agg_df,
        on="SK_ID_CURR",
        how="left"
    )

print("✅ Jointure finale terminée")
print("📐 Dimensions finales :", df_final.shape)
```

```python

# ============================================================
# CELLULE 8.9 - Sauvegarde du dataset final
# ============================================================

OUTPUT_PATH = PROCESSED_DIR / "e01_application_train_joined.csv"
df_final.to_csv(OUTPUT_PATH, index=False)

print(f"💾 Dataset final sauvegardé : {OUTPUT_PATH}")
```

```python

# ============================================================
# CELLULE 11.1 – FEATURE ENGINEERING
# ============================================================

import numpy as np
import pandas as pd

import gc

gc.collect()

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed"
)

print("🚀 Démarrage du Feature Engineering")

df_fe = df_final.copy()

# ------------------------------------------------------------------
# 1. Features métiers – Capacité de remboursement
# ------------------------------------------------------------------
# Ratio crédit / revenu
df_fe["CREDIT_INCOME_RATIO"] = (
    df_fe["AMT_CREDIT"] / df_fe["AMT_INCOME_TOTAL"]
)

# Ratio annuité / revenu
df_fe["ANNUITY_INCOME_RATIO"] = (
    df_fe["AMT_ANNUITY"] / df_fe["AMT_INCOME_TOTAL"]
)

# Ratio crédit / annuité
df_fe["CREDIT_ANNUITY_RATIO"] = (
    df_fe["AMT_CREDIT"] / df_fe["AMT_ANNUITY"]
)

# ------------------------------------------------------------------
# 2. Features liées à l’âge et à l’emploi
# ------------------------------------------------------------------
# Âge en années (DAYS_BIRTH est négatif)
df_fe["AGE_YEARS"] = (-df_fe["DAYS_BIRTH"]) / 365.25

# Ancienneté professionnelle en années
df_fe["EMPLOYMENT_YEARS"] = (
    -df_fe["DAYS_EMPLOYED"].replace(365243, np.nan)
) / 365.25

# Ratio ancienneté / âge
df_fe["EMPLOYMENT_AGE_RATIO"] = (
    df_fe["EMPLOYMENT_YEARS"] / df_fe["AGE_YEARS"]
)

# ------------------------------------------------------------------
# 3. Features liées à la famille
# ------------------------------------------------------------------
df_fe["INCOME_PER_PERSON"] = (
    df_fe["AMT_INCOME_TOTAL"] / df_fe["CNT_FAM_MEMBERS"]
)

# ------------------------------------------------------------------
# 4. Indicateurs simples de risque
# ------------------------------------------------------------------
df_fe["HAS_CHILDREN"] = (df_fe["CNT_CHILDREN"] > 0).astype(int)
df_fe["HAS_CAR"] = (df_fe["FLAG_OWN_CAR"] == "Y").astype(int)
df_fe["HAS_REALTY"] = (df_fe["FLAG_OWN_REALTY"] == "Y").astype(int)

# ------------------------------------------------------------------
# 5. Vérification des nouvelles features
# ------------------------------------------------------------------
new_features = list(set(df_fe.columns) - set(df_final.columns))

print(f"🧠 Nombre de nouvelles features créées : {len(new_features)}")
print("📌 Liste des nouvelles features :")
for feat in sorted(new_features):
    print(f" - {feat}")

# ------------------------------------------------------------------
# 6. Vérification des valeurs infinies ou aberrantes
# ------------------------------------------------------------------
inf_count = np.isinf(df_fe[new_features]).sum().sum()
print(f"\n⚠️ Valeurs infinies détectées : {inf_count}")

df_fe.replace([np.inf, -np.inf], np.nan, inplace=True)

# ------------------------------------------------------------------
# 7. Sauvegarde du dataset enrichi
# ------------------------------------------------------------------
output_path = PROCESSED_DIR / "e01_application_train_fe.csv"
df_fe.to_csv(output_path, index=False)

print(f"\n💾 Dataset enrichi sauvegardé dans : {output_path}")
print("✅ CELLULE 11 terminée avec succès")
```

```python

# ============================================================
# CELLULE 11.2 – ANALYSE DES CORRÉLATIONS, DOUBLONS & LEAKAGES
# ============================================================

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
import gc

gc.collect()

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (16, 10)

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed"
)

input_path = PROCESSED_DIR / "e01_application_train_fe.csv"

print("🚀 Chargement du dataset feature-engineered")
df_fe = pd.read_csv(input_path)

print(f"📐 Shape du dataset : {df_fe.shape}")
display(df_fe.head())

# ---------------------------------------
# 🔹 1. Analyse des doublons de lignes
# ---------------------------------------
print("\n🔍 Analyse des doublons de lignes")

nb_duplicates = df_fe.duplicated().sum()
print(f"➡️ Nombre de lignes dupliquées : {nb_duplicates}")

if nb_duplicates > 0:
    display(df_fe[df_fe.duplicated(keep=False)].head())


# ---------------------------------------
# 🔹 2. Analyse des colonnes strictement dupliquées
# ---------------------------------------
print("\n🔍 Analyse des colonnes dupliquées")

duplicated_cols_mask = df_fe.T.duplicated()
duplicated_cols = duplicated_cols_mask[duplicated_cols_mask].index.tolist()

print(f"➡️ Nombre de colonnes dupliquées : {len(duplicated_cols)}")

if len(duplicated_cols) > 0:
    print("📌 Colonnes concernées :")
    for col in duplicated_cols:
        print(f" - {col}")

# ---------------------------------------
# 🔹 3. Matrice de corrélation (variables numériques)
# ---------------------------------------
print("\n📊 Calcul de la matrice de corrélation")

df_num = df_fe.select_dtypes(include=["int64", "float64"])
corr_matrix = df_num.corr()

# ---------------------------------------
# 🔹 4. Heatmap des corrélations
# ---------------------------------------
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(18, 12))
sns.heatmap(
    corr_matrix,
    mask=mask,
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)
plt.title("Matrice de corrélation – Features numériques")
plt.show()

# ---------------------------------------
# 🔹 5. Détection des corrélations fortes (> 0.9)
# ---------------------------------------
print("\n🚨 Détection des corrélations fortes")

threshold = 0.9

high_corr_pairs = (
    corr_matrix.abs()
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

high_corr_pairs.columns = ["Feature_1", "Feature_2", "Correlation"]
high_corr_pairs = high_corr_pairs[
    high_corr_pairs["Correlation"] >= threshold
].sort_values(by="Correlation", ascending=False)

print(f"➡️ Nombre de paires avec corrélation ≥ {threshold} : {len(high_corr_pairs)}")
display(high_corr_pairs)

# ---------------------------------------
# 🔹 6. Analyse du risque de leakage via la cible
# ---------------------------------------
print("\n🧪 Analyse des corrélations avec la cible")

TARGET_COL = "TARGET"

if TARGET_COL in df_num.columns:
    target_corr = (
        corr_matrix[TARGET_COL]
        .abs()
        .sort_values(ascending=False)
    )

    display(
        target_corr.to_frame(name="|Corrélation avec TARGET|")
    )
else:
    print("⚠️ La colonne TARGET est absente du dataset.")
```

```python

# ============================================================
# CELLULE 12 – ENCODAGE (ORDINAL VS NOMINAL) + IMPUTATION
# ============================================================

import pandas as pd
from pathlib import Path
from sklearn.preprocessing import OneHotEncoder

import gc

gc.collect()

# ------------------------------------------------------------------
# 0. Dossier & fichier source
# ------------------------------------------------------------------
PROCESSED_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed"
)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

INPUT_PATH = PROCESSED_DIR / "e01_application_train_fe_no_leakage.csv"

print("🚀 Chargement du dataset feature-engineered sans leakage")
df_fe = pd.read_csv(INPUT_PATH)

print(f"📐 Shape du dataset : {df_fe.shape}")
display(df_fe.head())

# ------------------------------------------------------------------
# 1. Séparation des variables
# ------------------------------------------------------------------
df = df_fe.copy()

TARGET_COL = "TARGET"

numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()

if TARGET_COL in numerical_cols:
    numerical_cols.remove(TARGET_COL)

print(f"🔢 Variables numériques : {len(numerical_cols)}")
print(f"🔤 Variables catégorielles : {len(categorical_cols)}")

# ------------------------------------------------------------------
# 2. Définition des variables ordinales
# ------------------------------------------------------------------
ordinal_mappings = {
    "NAME_EDUCATION_TYPE": {
        "Lower secondary": 1,
        "Secondary / secondary special": 2,
        "Incomplete higher": 3,
        "Higher education": 4,
        "Academic degree": 5
    },
    "NAME_FAMILY_STATUS": {
        "Single / not married": 1,
        "Married": 2,
        "Civil marriage": 3,
        "Separated": 4,
        "Widow": 5
    }
}

ordinal_cols = [col for col in ordinal_mappings.keys() if col in df.columns]
nominal_cols = [col for col in categorical_cols if col not in ordinal_cols]

print(f"📐 Variables ordinales : {ordinal_cols}")
print(f"🏷️ Variables nominales : {len(nominal_cols)}")

# ------------------------------------------------------------------
# 3. Encodage ordinal sécurisé (NaN + catégories inconnues)
# ------------------------------------------------------------------
for col, mapping in ordinal_mappings.items():
    if col in df.columns:
        df[col] = (
            df[col]
            .map(mapping)
            .fillna(0)   # 0 = inconnu / non renseigné
            .astype("int64")
        )

print("✅ Encodage ordinal appliqué (sécurisé)")

# ------------------------------------------------------------------
# 4. Encodage nominal (One-Hot Encoding)
# ------------------------------------------------------------------
ohe = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore",
    drop=None
)

encoded_array = ohe.fit_transform(df[nominal_cols])
encoded_cols = ohe.get_feature_names_out(nominal_cols)

df_nominal_encoded = pd.DataFrame(
    encoded_array,
    columns=encoded_cols,
    index=df.index
)

print(f"✅ One-Hot Encoding appliqué : {df_nominal_encoded.shape[1]} nouvelles colonnes")

# ------------------------------------------------------------------
# 5. Reconstruction du DataFrame encodé
# ------------------------------------------------------------------
df_encoded = pd.concat(
    [
        df[numerical_cols + ordinal_cols],
        df_nominal_encoded,
        df[TARGET_COL]
    ],
    axis=1
)

print("📦 DataFrame encodé (avant imputation) :", df_encoded.shape)
print()

# ------------------------------------------------------------------
# 5.1 Vérification des colonnes avec NaN
# ------------------------------------------------------------------
na_cols = df_encoded.isna().sum()
na_cols = na_cols[na_cols > 0].sort_values(ascending=False)

print("Colonnes avec NaN après encodage :")
display(na_cols.head(15))

print()
print(
    "ℹ️ Les valeurs manquantes issues des tables secondaires "
    "représentent une absence d’historique client et non une donnée manquante classique."
)
print()

# ------------------------------------------------------------------
# 6. Imputation métier
# ------------------------------------------------------------------

# 6.1 Tables secondaires → 0 = absence d’historique
zero_impute_prefixes = ("CC_", "POS_", "BUREAU_", "PREV_")

zero_impute_cols = [
    col for col in df_encoded.columns
    if col.startswith(zero_impute_prefixes)
]

df_encoded[zero_impute_cols] = df_encoded[zero_impute_cols].fillna(0)

# 6.2 Colonnes restantes → médiane
remaining_na_cols = df_encoded.columns[df_encoded.isna().any()].tolist()

if remaining_na_cols:
    df_encoded[remaining_na_cols] = df_encoded[remaining_na_cols].fillna(
        df_encoded[remaining_na_cols].median()
    )

print("✅ Imputation métier appliquée")
print()

# ------------------------------------------------------------------
# 7. Vérifications finales
# ------------------------------------------------------------------
assert df_encoded.isna().sum().sum() == 0, "❌ Valeurs manquantes détectées après imputation"
assert TARGET_COL in df_encoded.columns, "❌ TARGET absente après encodage"

print("✅ Vérifications OK – dataset prêt pour la modélisation")
print()

# ------------------------------------------------------------------
# 8. Sauvegarde
# ------------------------------------------------------------------
output_path = PROCESSED_DIR / "e01_application_train_encoded.csv"
df_encoded.to_csv(output_path, index=False)

print(f"💾 Dataset encodé sauvegardé : {output_path}")
print("✅ CELLULE 12 terminée avec succès")
```

```python

# ============================================================
# CELLULE 1.3 – RECHARGEMENT DES DATASETS (ISSUS ÉTAPE 1)
# ============================================================

import pandas as pd
from pathlib import Path
import mlflow
import hashlib
import gc

gc.collect()

# ------------------------------------------------------------------
# 1. Chemin local dataset (notebook uniquement)
# ------------------------------------------------------------------
DATA_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/"
    "Projet_06_Initiez_vous_au_MLOps_partie_1/data/processed/parquet"
)

assert DATA_DIR.exists(), f"❌ Dossier introuvable : {DATA_DIR}"

# ------------------------------------------------------------------
# 2. 2. Fichiers versionnés (Étape 1) + Préfixes des datasets (logiques, pas de timestamp hardcodé)
# ------------------------------------------------------------------
DATASET_PREFIXES = {
    "application_train_clean": "e01_application_train_clean_",
    "application_train_joined": "e01_application_train_joined_",
    "application_train_fe_no_leakage": "e01_application_train_fe_no_leakage_",
    "application_train_encoded": "e01_application_train_encoded_",
    "application_test_clean": "e01_application_test_clean_",
}

# ------------------------------------------------------------------
# 3. Fonction MLOps pour récupérer la dernière version
# ------------------------------------------------------------------
def find_latest_parquet(prefix: str, directory: Path) -> Path:
    candidates = list(directory.glob(f"{prefix}*.parquet"))
    assert len(candidates) > 0, f"Aucun fichier trouvé pour {prefix}"

    # tri par timestamp dans le nom
    candidates.sort(key=lambda p: p.stem.split("_v")[-1], reverse=True)
    return candidates[0]

# ------------------------------------------------------------------
# 4. Chargement + fingerprint
# ------------------------------------------------------------------
datasets = {}
dataset_fingerprints = {}
dataset_files_used = {}

for name, prefix in DATASET_PREFIXES.items():
    path = find_latest_parquet(prefix, DATA_DIR)

    df = pd.read_parquet(path)
    datasets[name] = df
    dataset_files_used[name] = path.name

    # fingerprint structurel
    signature = f"{df.shape[0]}x{df.shape[1]}|" + ",".join(df.columns)
    fingerprint = hashlib.md5(signature.encode()).hexdigest()
    dataset_fingerprints[name] = fingerprint

    print(f"✅ {name}")
    print(f"   ↳ fichier : {path.name}")
    print(f"   ↳ shape   : {df.shape}")
    print(f"   ↳ fp      : {fingerprint[:8]}")

# ------------------------------------------------------------------
# 4. Log MLflow (métadonnées seulement)
# ------------------------------------------------------------------
for name, fp in dataset_fingerprints.items():
    mlflow.log_param(f"dataset_{name}_fingerprint", fp)
    mlflow.log_param(f"dataset_{name}_file", dataset_files_used[name])

mlflow.log_param("dataset_source", "Colab Drive – Étape 1 parquet (latest)")

print("\n📦 Datasets disponibles :", list(datasets.keys()))
```

```python

# ============================================================
# CELLULE 1 – IMPORTS & CONFIGURATION (ÉTAPE 3)
# Modélisation multi-algorithmes + CV stratifiée
# Tracking MLflow centralisé (HF Space)
# ============================================================

# -----------------------------
# Librairies standards
# -----------------------------
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Cross-validation & métriques
# -----------------------------
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import (
    roc_auc_score,
    recall_score,
    f1_score,
    make_scorer
)

# -----------------------------
# Modèles Scikit-learn
# -----------------------------
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier

# -----------------------------
# Boosting (compatibles MLflow HF Space)
# -----------------------------
import xgboost as xgb
import lightgbm as lgb

# -----------------------------
# MLflow (tracking centralisé)
# -----------------------------
import mlflow
import mlflow.sklearn

import gc

gc.collect()

# ============================================================
# CONFIGURATION MLFLOW (HF Space)
# ============================================================

MLFLOW_TRACKING_URI = "https://remdev-ai-projet-06-mlops-mlflow.hf.space"

mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
mlflow.set_experiment("Projet_06_MLOps_Step_03_Model_Comparison")

print(f"✅ MLflow connecté à : {MLFLOW_TRACKING_URI}")
print("🧪 Expérience : Projet_06_MLOps_Step_03_Model_Comparison")

# Désactiver l'autolog global (on gère tout proprement)
mlflow.autolog(disable=True)

# ============================================================
# CONFIGURATION GLOBALE DE L'ÉVALUATION
# ============================================================

RANDOM_STATE = 42
N_SPLITS = 5

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

print(f"🔁 Cross-validation : StratifiedKFold(n_splits={N_SPLITS})")

# ============================================================
# MÉTRIQUE MÉTIER (FN >> FP)
# ============================================================

# Hypothèse métier :
# - FN = client à risque accepté → coût élevé
# - FP = client solvable refusé → coût faible
# FN pénalisé 5x plus que FP

def business_cost(y_true, y_pred):
    fn = np.sum((y_true == 1) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    return -(5 * fn + fp)  # score à MAXIMISER

business_scorer = make_scorer(
    business_cost,
    greater_is_better=True
)

# ============================================================
# MÉTRIQUES UTILISÉES POUR LA COMPARAISON DES MODÈLES
# ============================================================

scoring = {
    "auc": "roc_auc",
    "recall_minority": make_scorer(recall_score, pos_label=1),
    "f1": make_scorer(f1_score, pos_label=1),
    "business": business_scorer
}

print("📊 Métriques : AUC | Recall classe minoritaire | F1 | Business Cost")
print("🚀 Étape 3 – Prêt pour l'évaluation multi-modèles")
```

```python

# ============================================================
# CELLULE 3 – MODÈLES TESTÉS (ÉTAPE 3)
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
import xgboost as xgb
import lightgbm as lgb
import gc

gc.collect()

# ------------------------------------------------------------
# Ratio de déséquilibre (classe majoritaire / minoritaire)
# Calculé en Cellule 2
# ------------------------------------------------------------
assert "imbalance_ratio" in globals(), "❌ imbalance_ratio non défini (Cellule 2 manquante)"

pos_weight = imbalance_ratio

# ------------------------------------------------------------
# Définition des modèles comparés — MODE DEV
# ------------------------------------------------------------
models = {

    # ------------------------------------
    # Logistic Regression (baseline ultra rapide)
    # ------------------------------------
    "LogisticRegression": Pipeline(steps=[
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(
            max_iter=300,                 # 2000 → 300
            class_weight="balanced",
            solver="lbfgs",
            n_jobs=1,
            random_state=RANDOM_STATE
        ))
    ]),

    # ------------------------------------
    # Random Forest (beaucoup plus petit)
    # ------------------------------------
    "RandomForest": RandomForestClassifier(
        n_estimators=50,                  # 200 → 50
        max_depth=8,                      # 15 → 8
        min_samples_leaf=200,             # 50 → 200 (moins d’arbres profonds)
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=1
    ),

    # ------------------------------------
    # XGBoost (DEV fast)
    # ------------------------------------
    "XGBoost": xgb.XGBClassifier(
        eval_metric="auc",
        scale_pos_weight=pos_weight,
        n_estimators=80,                  # 300 → 80
        max_depth=4,                      # 6 → 4
        learning_rate=0.1,                # 0.05 → 0.1 (converge plus vite)
        subsample=0.7,
        colsample_bytree=0.7,
        tree_method="hist",
        n_jobs=1,
        random_state=RANDOM_STATE
    ),

    # ------------------------------------
    # LightGBM (DEV fast)
    # ------------------------------------
    "LightGBM": lgb.LGBMClassifier(
        class_weight="balanced",
        n_estimators=100,                 # 300 → 100
        max_depth=6,                      # 10 → 6
        num_leaves=16,                    # 31 → 16
        subsample=0.7,
        colsample_bytree=0.7,
        random_state=RANDOM_STATE,
        n_jobs=1
    ),

    # ------------------------------------
    # MLP (beaucoup plus petit)
    # ------------------------------------
    "MLP": MLPClassifier(
        hidden_layer_sizes=(32, 16),       # (64,32) → (32,16)
        max_iter=100,                     # 300 → 100
        early_stopping=True,
        random_state=RANDOM_STATE
    )
}

print(f"✅ {len(models)} modèles prêts pour la comparaison MLflow")
```

```python

# ============================================================
# CELLULE 4 – ENTRAÎNEMENT MULTI-MODÈLES + CV STRATIFIÉE
# Tracking MLflow – Étape 3
# ============================================================

import gc
gc.collect()

# Sécurité
assert "models" in globals()
assert "cv" in globals()
assert "scoring" in globals()

results = []

for model_name, model in models.items():

    print(f"\n🚀 Entraînement du modèle : {model_name}")

    with mlflow.start_run(run_name=f"step3_{model_name}"):

        # ----------------------------------------------------
        # Tags MLOps (essentiels pour le dashboard HF)
        # ----------------------------------------------------
        mlflow.set_tags({
            "step": "03_model_comparison",
            "model_name": model_name,
            "cv_folds": N_SPLITS,
            "metric_optim": "business_cost",
            "dataset": "application_train_encoded",
            "tracking_backend": "mlflow_hf_space"
        })

        # ----------------------------------------------------
        # Cross-validation stratifiée
        # ----------------------------------------------------
        cv_results = cross_validate(
            model,
            X,
            y,
            cv=cv,
            scoring=scoring,
            return_train_score=False,
            n_jobs=1
        )

        # ----------------------------------------------------
        # Agrégation des métriques (moyenne CV)
        # ----------------------------------------------------
        metrics = {
            "val_auc": float(np.mean(cv_results["test_auc"])),
            "val_recall_minority": float(np.mean(cv_results["test_recall_minority"])),
            "val_f1": float(np.mean(cv_results["test_f1"])),
            "val_business_cost": float(np.mean(cv_results["test_business"])),
        }

        results.append({
            "model": model_name,
            **metrics
        })

        # ----------------------------------------------------
        # Logging MLflow
        # ----------------------------------------------------
        for k, v in metrics.items():
            mlflow.log_metric(k, v)

        # Paramètres (filtrage pour MLflow HF)
        params = {
            k: v for k, v in model.get_params().items()
            if isinstance(v, (int, float, str, bool))
        }
        mlflow.log_params(params)

        print("📊 Résultats CV :")
        print(metrics)

print("\n✅ Entraînement multi-modèles terminé")
```

```python

# ============================================================
# CELLULE 3 – MÉTRIQUE MÉTIER (VERSION SEUIL + PROBA)
# Compatible Optuna / GridSearchCV / MLflow
# ============================================================

import numpy as np
from sklearn.metrics import confusion_matrix
import gc

gc.collect()

# ------------------------------------------------------------------
# Pondérations métier
# ------------------------------------------------------------------
FN_COST = 5   # client risqué accepté
FP_COST = 1   # client sain refusé

print(f"⚖️ Coûts métier : FN={FN_COST} | FP={FP_COST}")

# ------------------------------------------------------------------
# Conversion proba → prédiction selon seuil
# ------------------------------------------------------------------
def proba_to_pred(y_proba, threshold):
    return (y_proba >= threshold).astype(int)

# ------------------------------------------------------------------
# Coût métier basé sur probabilités + seuil
# ------------------------------------------------------------------
def business_cost_from_proba(y_true, y_proba, threshold):
    y_pred = proba_to_pred(y_proba, threshold)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    return FN_COST * fn + FP_COST * fp

# ------------------------------------------------------------------
# Score métier à MAXIMISER (pour Optuna / MLflow)
# ------------------------------------------------------------------
def business_score_from_proba(y_true, y_proba, threshold):
    return -business_cost_from_proba(y_true, y_proba, threshold)
```

### Cellule 7 - Interprétation métier

- Le modèle prédit la probabilité de défaut du client.
- Cette probabilité est convertie en **score crédit** :
  
  Score = 1 − P(défaut)

- La décision de crédit est prise à l’aide du **seuil métier optimisé** qui minimise le coût des erreurs (FN ≫ FP).


```python

# ============================================================
# CELLULE 7 – SCORE CRÉDIT & DÉCISION MÉTIER (FINAL)
# Alignée avec Optuna + seuil métier
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_predict
import gc

gc.collect()

# ------------------------------------------------------------
# 1. Probabilités out-of-fold du champion MLflow
# ------------------------------------------------------------
y_proba_default = cross_val_predict(
    champion_model,
    X,
    y,
    cv=cv,
    method="predict_proba",
    n_jobs=1
)[:, 1]

# ------------------------------------------------------------
# 2. Application STRICTE du seuil métier optimisé
# ------------------------------------------------------------
y_pred_optimal = proba_to_pred(y_proba_default, best_threshold)

# ------------------------------------------------------------
# 3. Score crédit (métier)
# Score = probabilité de remboursement
# ------------------------------------------------------------
credit_score = 1 - y_proba_default

# ------------------------------------------------------------
# 4. Décision métier finale
# ------------------------------------------------------------
credit_decision = np.where(
    y_pred_optimal == 0,
    "Crédit accordé",
    "Crédit refusé"
)

# ------------------------------------------------------------
# 5. DataFrame de scoring final
# ------------------------------------------------------------
df_scoring = pd.DataFrame({
    "Probabilité de défaut": y_proba_default,
    "Score crédit (P remboursement)": credit_score,
    "Décision": credit_decision,
    "Vrai statut": y.map({0: "Client solvable", 1: "Client en défaut"})
})

print("✅ Score crédit & décisions métier générés à partir du champion MLflow")
df_scoring.head(10)
```
