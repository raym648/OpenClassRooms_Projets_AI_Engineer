```python

# ==================================================
# Cellule 5 - Création de l'environnement RL
# ==================================================

import gymnasium as gym

# Instanciation de l'environnement (API Gymnasium)
env = gym.make("CartPole-v1")

print("✅ Environnement créé :", env)
```

```python

# ==================================================
# Cellule 6 - Inspection des espaces (version enrichie)
# ==================================================

import numpy as np

# ================================
# Observation space (état)
# ================================
print("📊 Observation space :")
print(env.observation_space)

obs_space = env.observation_space

print("\n🔍 Observation space - Analyse détaillée :")

# Récupération des bornes
low = obs_space.low
high = obs_space.high

# Fonction utilitaire pour afficher inf proprement
def format_bound(value):
    if np.isinf(value):
        return "∞" if value > 0 else "-∞"
    return f"{value:.4f}"

# Décomposition des 4 variables
labels = [
    "Position du chariot",
    "Vitesse du chariot",
    "Angle du poteau",
    "Vitesse angulaire du poteau"
]

print("\n🧠 Décomposition précise :\n")

for i, label in enumerate(labels):
    min_val = format_bound(low[i])
    max_val = format_bound(high[i])

    print(f"{i+1}. {label}")
    print(f"   bornes : [{min_val}, {max_val}]")

    # Ajout d'interprétation spécifique
    if i == 2:
        print("   ≈ ±24° (cohérence physique)")

    if "∞" in min_val or "∞" in max_val:
        print("   (non borné)")

    print()

# ================================
# Action space (actions)
# ================================
print("🎮 Action space :")
print(env.action_space)

print("\n🧠 Interprétation :\n")

print("Nombre d'actions possibles :", env.action_space.n)
print("0 → gauche")
print("1 → droite")
```

```python

# ==================================================
# Cellule 7 - Analyse technique des espaces
# ==================================================

# --- Observation space ---
obs_space = env.observation_space

print("🔍 Observation space détails :")
print("Type :", type(obs_space))
print("Shape :", obs_space.shape)
print("Bornes min :", obs_space.low)
print("Bornes max :", obs_space.high)

# --- Action space ---
action_space = env.action_space

print("\n🔍 Action space détails :")
print("Type :", type(action_space))
print("Nombre d'actions :", action_space.n)
```

```python

# ==================================================
# Cellule 8 - Sampling (données aléatoires enrichi)
# ==================================================

import numpy as np

# Génère une observation aléatoire (vecteur continu)
random_observation = env.observation_space.sample()

# Génère une action aléatoire (discrète)
random_action = env.action_space.sample()

print("🎲 Observation aléatoire :", random_observation)
print("🎲 Action aléatoire :", random_action)

# ================================
# Analyse de l'observation
# ================================
print("\n🔍 Analyse de l'observation aléatoire :\n")

labels = [
    "Position du chariot",
    "Vitesse du chariot",
    "Angle du poteau (radians)",
    "Vitesse angulaire du poteau"
]

for i, value in enumerate(random_observation):
    print(f"{i+1}. {labels[i]} : {value:.6f}")

    # Interprétation spécifique pour l'angle
    if i == 2:
        angle_deg = np.degrees(value)
        print(f"   → ≈ {angle_deg:.2f}°")

    print()

# ================================
# Analyse de l'action
# ================================
print("🎮 Interprétation de l'action :\n")

if random_action == 0:
    print("Action = 0 → Déplacement du chariot vers la GAUCHE")
elif random_action == 1:
    print("Action = 1 → Déplacement du chariot vers la DROITE")
else:
    print("⚠️ Action inattendue :", random_action)

# ================================
# Rappel conceptuel
# ================================
print("\n🧠 Interprétation RL :")
print("→ L'observation représente l'état courant de l'environnement")
print("→ L'action est une décision prise sans stratégie (aléatoire)")
print("→ Ce mécanisme simule une phase d'exploration brute")
```

```python

# ==================================================
# Cellule 9 - Première interaction avec l'environnement (version enrichie)
# ==================================================

import numpy as np

# ================================
# 1. RESET (nouvel épisode)
# ================================
observation, info = env.reset()

print("🚀 Observation initiale :", observation)

print("\n🔍 Analyse de l'observation initiale :\n")

labels = [
    "Position du chariot",
    "Vitesse du chariot",
    "Angle du poteau (radians)",
    "Vitesse angulaire du poteau"
]

for i, value in enumerate(observation):
    print(f"{i+1}. {labels[i]} : {value:.6f}")

    if i == 2:
        angle_deg = np.degrees(value)
        print(f"   → ≈ {angle_deg:.2f}° (proche de l'équilibre)")

    print()

# ================================
# 2. ACTION
# ================================
action = env.action_space.sample()

print("🎮 Action prise :", action)

if action == 0:
    print("→ Interprétation : déplacement vers la GAUCHE")
elif action == 1:
    print("→ Interprétation : déplacement vers la DROITE")
else:
    print("⚠️ Action inattendue")

# ================================
# 3. STEP (transition)
# ================================
next_observation, reward, terminated, truncated, info = env.step(action)

print("\n🔄 Nouvelle observation :", next_observation)

print("\n🔍 Analyse de la transition :\n")

for i, value in enumerate(next_observation):
    delta = value - observation[i]

    print(f"{i+1}. {labels[i]} : {value:.6f}")
    print(f"   Variation : {delta:+.6f}")

    if i == 2:
        angle_deg = np.degrees(value)
        print(f"   → ≈ {angle_deg:.2f}°")

    print()

# ================================
# 4. REWARD
# ================================
print("🎯 Récompense :", reward)

if reward == 1.0:
    print("→ Interprétation : le poteau est toujours en équilibre")
else:
    print("→ Récompense atypique (à analyser)")

# ================================
# 5. TERMINATED / TRUNCATED
# ================================
print("\n🛑 État de l'épisode :")
print("Terminated :", terminated)
print("Truncated :", truncated)

if not terminated and not truncated:
    print("→ L'épisode continue (état valide)")
elif terminated:
    print("→ Échec : le poteau est tombé ou condition terminale atteinte")
elif truncated:
    print("→ Fin d'épisode : limite de temps atteinte (TimeLimit wrapper)")

# ================================
# 6. RÉSUMÉ DU CYCLE RL
# ================================
print("\n🧠 Cycle RL observé :")
print("Observation → Action → Nouvelle observation → Récompense")
```

```python

# ==================================================
# Cellule 2 - Imports et configuration globale
# ==================================================

import gymnasium as gym  # Bibliothèque RL standard
import os                # Gestion des chemins fichiers
import json              # Sauvegarde des résultats

# Définition du chemin racine du projet (Google Drive)
PROJECT_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement"

# Chemin du pyproject.toml (source de vérité des dépendances)
PYPROJECT_PATH = os.path.join(PROJECT_PATH, "Configs", "pyproject.toml")

# Dossier spécifique à l'exercice 1
EXERCICE_PATH = os.path.join(PROJECT_PATH, "Exercice-01")

# Création du dossier si inexistant (sécurité)
os.makedirs(EXERCICE_PATH, exist_ok=True)

print("📁 Dossier de travail :", EXERCICE_PATH)
```

```python

# ==================================================
# Cellule 3 - Création de l'environnement CartPole
# ==================================================

# Création de l'environnement
env = gym.make("CartPole-v1")

print("✅ Environnement créé :", env)
```

```python

# ==================================================
# Cellule 4 - Simulation avec politique aléatoire (boucle RL complète) avec affichage détaillé
# ==================================================

NUM_EPISODES = 10  # ⚠️ réduit pour lisibilité
MAX_STEPS_TO_PRINT = 10

episode_rewards = []

for episode in range(NUM_EPISODES):

    print(f"\n==================== ÉPISODE {episode + 1} ====================")

    # --- 1. Observation initiale ---
    observation, info = env.reset()

    x, x_dot, theta, theta_dot = observation

    print(f"🟢 Observation initiale : {observation}")
    print(f"   → Position chariot        (x)        : {x:.4f}")
    print(f"   → Vitesse chariot         (x_dot)    : {x_dot:.4f}")
    print(f"   → Angle du poteau         (theta)    : {theta:.4f}")
    print(f"   → Vitesse angulaire       (theta_dot): {theta_dot:.4f}")

    total_reward = 0
    terminated = False
    truncated = False
    step_count = 0

    while not (terminated or truncated):

        step_count += 1

        # --- 2. Action prise ---
        action = env.action_space.sample()

        action_meaning = "Droite ➡️" if action == 1 else "Gauche ⬅️"

        # --- Interaction environnement ---
        new_observation, reward, terminated, truncated, info = env.step(action)

        nx, nx_dot, ntheta, ntheta_dot = new_observation

        total_reward += reward

        # --- Affichage détaillé ---
        if step_count <= MAX_STEPS_TO_PRINT:
            print(f"\n--- Step {step_count} ---")

            print(f"🎮 Action prise : {action} ({action_meaning})")
            print(f"   → Interprétation : Force appliquée au chariot")

            print(f"🔄 Nouvelle observation : {new_observation}")
            print(f"   → x        : {nx:.4f} (position)")
            print(f"   → x_dot    : {nx_dot:.4f} (vitesse)")
            print(f"   → theta    : {ntheta:.4f} (angle)")
            print(f"   → theta_dot: {ntheta_dot:.4f} (vitesse angulaire)")

            print(f"🎯 Récompense : {reward}")
            print(f"   → Interprétation : +1 tant que le poteau tient")

            print(f"🛑 Terminated : {terminated}")
            print(f"⏱️ Truncated  : {truncated}")

            if terminated:
                print("   → Fin : le poteau est tombé ❌")
            elif truncated:
                print("   → Fin : limite de temps atteinte ⏱️")
            else:
                print("   → État : système encore stable")

        # Mise à jour état
        observation = new_observation

    episode_rewards.append(total_reward)

    print(f"\n🎯 Reward total épisode {episode + 1} : {total_reward}")
    print(f"📊 Nombre total de steps             : {step_count}")

    # --- Synthèse épisode ---
    if total_reward < 20:
        print("📉 Synthèse : échec rapide (instabilité forte)")
    elif total_reward < 50:
        print("📊 Synthèse : performance moyenne (équilibre instable)")
    else:
        print("📈 Synthèse : bonne performance aléatoire (séquence favorable)")

# Fermeture environnement
env.close()

# ==================================================
# Synthèse globale
# ==================================================

import numpy as np

rewards_array = np.array(episode_rewards)

print("\n==================== SYNTHÈSE GLOBALE ====================")
print(f"📊 Nombre d'épisodes : {len(rewards_array)}")
print(f"📊 Rewards : {episode_rewards}")
print(f"📊 Reward Moyenne : {rewards_array.mean():.2f}")
print(f"📊 Reward Min     : {rewards_array.min()}")
print(f"📊 Reward Max     : {rewards_array.max()}")
print(f"📊 Écart-type       : {rewards_array.std():.2f}")

# Interprétation globale
if rewards_array.mean() < 20:
    print("📉 Interprétation globale : agent très instable")
elif rewards_array.mean() < 40:
    print("📊 Interprétation globale : comportement aléatoire typique")
else:
    print("📈 Interprétation globale : résultats anormalement élevés (à vérifier)")

print("\n✅ Simulation terminée")
```

```python

# ==================================================
# Cellule 6 - Sauvegarde des résultats
# ==================================================

# Chemin du fichier de sortie
output_path = os.path.join(EXERCICE_PATH, "ex01_e02_random_policy_results.json")

# Structure des données à sauvegarder
results = {
    "experiment": "random_policy_cartpole",
    "num_episodes": len(episode_rewards),
    "episode_rewards": episode_rewards
}

# Sauvegarde en JSON
with open(output_path, "w") as f:
    json.dump(results, f, indent=4)

print("💾 Résultats sauvegardés :", output_path)
```

## 🧭 Étape 2 : Implémentez la boucle de Q-Learning

```python

# ==================================================
# Cellule 1 - Chargement environnement + Q-table initiale
# ==================================================

import numpy as np
import gymnasium as gym
import os

# Chemin racine du projet
PROJECT_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement"

# Chemin vers la Q-table initiale
QTABLE_PATH = os.path.join(
    PROJECT_PATH,
    "Exercice-02",
    "ex02_e01_q_table_initial.npy"
)

# Création de l'environnement FrozenLake
env = gym.make("FrozenLake-v1")

print("✅ Environnement créé :", env)

print()

# Chargement de la Q-table initiale
q_table = np.load(QTABLE_PATH)

print("✅ Q-table chargée :", q_table.shape)

# Vérification des dimensions
state_size = env.observation_space.n
action_size = env.action_space.n

print("États :", state_size, "| Actions :", action_size)
```

```python

# ==================================================
# Cellule 2 - Hyperparamètres Q-Learning
# ==================================================

# Nombre d'épisodes
NUM_EPISODES = 5000

# Nombre maximum d'étapes par épisode
MAX_STEPS = 100

# Taux d'apprentissage (alpha)
alpha = 0.1

# Facteur de discount (gamma)
gamma = 0.99

# Paramètres epsilon-greedy
epsilon = 1.0        # exploration initiale
epsilon_min = 0.01   # minimum
epsilon_decay = 0.995  # décroissance

# Stockage des métriques
rewards_per_episode = []
epsilon_history = []
```

```python

# ==================================================
# Cellule 3 - Boucle de Q-Learning
# ==================================================

import random

for episode in range(NUM_EPISODES):

    # Réinitialisation de l'environnement
    state, _ = env.reset()

    total_rewards = 0

    for step in range(MAX_STEPS):

        # =========================
        # 1. Choix de l'action (epsilon-greedy)
        # =========================
        if random.uniform(0, 1) < epsilon:
            # Exploration : action aléatoire
            action = env.action_space.sample()
        else:
            # Exploitation : meilleure action connue
            action = np.argmax(q_table[state, :])

        # =========================
        # 2. Exécution de l'action
        # =========================
        new_state, reward, terminated, truncated, _ = env.step(action)

        # =========================
        # 3. Mise à jour Bellman
        # =========================

        # Valeur actuelle
        old_value = q_table[state, action]

        # Meilleure valeur future (ATTENTION : new_state)
        next_max = np.max(q_table[new_state, :])

        # Formule de Bellman (mise à jour incrémentale)
        new_value = old_value + alpha * (reward + gamma * next_max - old_value)

        # Mise à jour ciblée
        q_table[state, action] = new_value

        # =========================
        # 4. Mise à jour état
        # =========================
        state = new_state
        total_rewards += reward

        # Fin d’épisode
        if terminated or truncated:
            break

    # =========================
    # 5. Décroissance de epsilon
    # =========================
    epsilon = max(epsilon_min, epsilon * epsilon_decay)

    # Stockage métriques
    rewards_per_episode.append(total_rewards)
    epsilon_history.append(epsilon)

    # Logging léger
    if (episode + 1) % 500 == 0:
        print(f"Episode {episode+1} | Reward moyen: {np.mean(rewards_per_episode[-500:]):.4f} | Epsilon: {epsilon:.4f}")
```

```python

# ==================================================
# Cellule 4 - Paramètres d’évaluation
# ==================================================

NUM_EPISODES = 1000  # Nombre d’épisodes d’évaluation
MAX_STEPS = 100     # Limite pour éviter boucle infinie

total_wins = 0
episode_results = []
```

```python

# ==================================================
# Cellule 5 - Boucle d’évaluation (100% exploitation)
# ==================================================

for episode in range(NUM_EPISODES):

    state, _ = env.reset()
    done = False
    step_count = 0

    while not done and step_count < MAX_STEPS:

        # Politique gloutonne (aucune exploration)
        action = np.argmax(q_table[state, :])

        next_state, reward, terminated, truncated, _ = env.step(action)

        done = terminated or truncated
        state = next_state
        step_count += 1

    # Succès si récompense finale = 1
    success = 1 if reward == 1.0 else 0
    total_wins += success

    episode_results.append({
        "episode": episode,
        "success": success,
        "steps": step_count
    })
```

```python

# ==================================================
# Cellule 6 - Calcul des métriques globales
# ==================================================

success_rate = (total_wins / NUM_EPISODES) * 100
avg_steps = np.mean([ep["steps"] for ep in episode_results])

print("\n📊 Résultats de l’évaluation")
print(f"Nombre d’épisodes : {NUM_EPISODES}")
print(f"Succès : {total_wins}")
print(f"Taux de réussite : {success_rate:.2f}%")
print(f"Longueur moyenne des épisodes : {avg_steps:.2f}")
```

```python

# ==================================================
# Cellule 3 - Définition du modèle DQN (sauvegarde fichier)
# ==================================================

dqn_code = '''
import torch
import torch.nn as nn
import torch.nn.functional as F

class DQN(nn.Module):
    def __init__(self, state_dim, action_dim):
        super(DQN, self).__init__()

        # Couche d'entrée -> première couche cachée
        self.fc1 = nn.Linear(state_dim, 128)

        # Deuxième couche cachée
        self.fc2 = nn.Linear(128, 128)

        # Couche de sortie -> Q-values pour chaque action
        self.fc3 = nn.Linear(128, action_dim)

    def forward(self, x):
        """
        Forward pass :
        entrée = état
        sortie = Q-values pour chaque action
        """

        # Activation ReLU après chaque couche cachée
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))

        # Pas d'activation finale (régression)
        return self.fc3(x)
'''

# Sauvegarde du fichier
dqn_path = os.path.join(EX3_PATH, "ex03_e01_dqn_model.py")

with open(dqn_path, "w") as f:
    f.write(dqn_code)

print("💾 Modèle DQN sauvegardé :", dqn_path)
```

```python

# ==================================================
# Cellule 4 - Replay Buffer (sauvegarde fichier)
# ==================================================

buffer_code = '''
import random
from collections import deque

class ReplayBuffer:
    def __init__(self, capacity):
        """
        capacity : nombre maximum d'expériences stockées
        """
        self.buffer = deque(maxlen=capacity)

    def push(self, state, action, reward, next_state, done):
        """
        Ajoute une transition dans le buffer
        """
        self.buffer.append((state, action, reward, next_state, done))

    def sample(self, batch_size):
        """
        Échantillonnage aléatoire d'un batch
        """
        batch = random.sample(self.buffer, batch_size)

        states, actions, rewards, next_states, dones = zip(*batch)

        return states, actions, rewards, next_states, dones

    def __len__(self):
        """
        Retourne la taille actuelle du buffer
        """
        return len(self.buffer)
'''

# Sauvegarde
buffer_path = os.path.join(EX3_PATH, "ex03_e01_replay_buffer.py")

with open(buffer_path, "w") as f:
    f.write(buffer_code)

print("💾 ReplayBuffer sauvegardé :", buffer_path)
```

```python

# ==================================================
# Cellule 1 - Chemins projet (respect structure imposée)
# ==================================================

import os
import random
import math
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim

import gymnasium as gym


PROJECT_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement"
EX03_PATH = os.path.join(PROJECT_PATH, "Exercice-03")

# Ajout au PYTHONPATH pour importer les modules existants
import sys
sys.path.append(EX03_PATH)

# Import des fichiers existants
from ex03_e01_replay_buffer import ReplayBuffer
from ex03_e01_dqn_model import DQN
```

```python

# ==================================================
# Cellule 2 - Initialisation environnement CartPole
# ==================================================

# Création de l'environnement CartPole-v1
env = gym.make("CartPole-v1")

print("✅ Environnement créé :", env)

print()

state_dim = env.observation_space.shape[0]
action_dim = env.action_space.n

print("📊 State dim :", state_dim)
print("🎮 Action dim :", action_dim)
```

```python

# ==================================================
# Cellule 3 - Hyperparamètres DQN
# ==================================================

BATCH_SIZE = 64
GAMMA = 0.99
LR = 1e-3

EPS_START = 1.0
EPS_END = 0.05
EPS_DECAY = 500

TARGET_UPDATE = 10  # en épisodes

NUM_EPISODES = 500
MAX_STEPS = 500

REPLAY_BUFFER_CAPACITY = 10000
```

```python

# ==================================================
# Cellule 4 - Initialisation réseaux + optimizer
# ==================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

policy_net = DQN(state_dim, action_dim).to(device)
target_net = DQN(state_dim, action_dim).to(device)

# Initialisation target = policy
target_net.load_state_dict(policy_net.state_dict())
target_net.eval()

optimizer = optim.Adam(policy_net.parameters(), lr=LR)

replay_buffer = ReplayBuffer(REPLAY_BUFFER_CAPACITY)
```

```python

# ==================================================
# Cellule 5 - Fonction epsilon-greedy
# ==================================================

def select_action(state, steps_done):
    """
    Sélection d'action avec stratégie epsilon-greedy
    """

    epsilon = EPS_END + (EPS_START - EPS_END) * math.exp(-1. * steps_done / EPS_DECAY)

    if random.random() < epsilon:
        return env.action_space.sample()
    else:
        with torch.no_grad():
            state_tensor = torch.FloatTensor(state).unsqueeze(0).to(device)
            q_values = policy_net(state_tensor)
            return q_values.argmax().item()
```

```python

# ==================================================
# Cellule 6 - Fonction optimize_model (cœur du DQN)
# ==================================================

def optimize_model():
    """
    Une étape d'apprentissage DQN :
    - sampling batch
    - calcul Q(s,a)
    - calcul target
    - loss + backprop
    """

    if len(replay_buffer) < BATCH_SIZE:
        return None

    # Sample batch
    states, actions, rewards, next_states, dones = replay_buffer.sample(BATCH_SIZE)

    # Conversion en tensors
    states = torch.FloatTensor(states).to(device)
    actions = torch.LongTensor(actions).unsqueeze(1).to(device)
    rewards = torch.FloatTensor(rewards).to(device)
    next_states = torch.FloatTensor(next_states).to(device)
    dones = torch.FloatTensor(dones).to(device)

    # -------------------------------
    # Q(s,a) via policy_net
    # -------------------------------
    q_values = policy_net(states)
    state_action_values = q_values.gather(1, actions).squeeze(1)

    # -------------------------------
    # max_a Q_target(s',a)
    # -------------------------------
    with torch.no_grad():
        next_q_values = target_net(next_states)
        max_next_q_values = next_q_values.max(1)[0]

    # -------------------------------
    # Calcul target
    # -------------------------------
    expected_q_values = rewards + (GAMMA * max_next_q_values * (1 - dones))

    # -------------------------------
    # Loss
    # -------------------------------
    loss = nn.MSELoss()(state_action_values, expected_q_values)

    # -------------------------------
    # Backpropagation
    # -------------------------------
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    return loss.item()
```

```python

# ==================================================
# Cellule 7 - Boucle principale d'entraînement
# ==================================================

rewards_history = []
loss_history = []

steps_done = 0

for episode in range(NUM_EPISODES):

    state, _ = env.reset()
    total_reward = 0

    for step in range(MAX_STEPS):

        # Sélection action
        action = select_action(state, steps_done)
        steps_done += 1

        # Interaction environnement
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        # Stockage transition
        replay_buffer.push(state, action, reward, next_state, done)

        state = next_state
        total_reward += reward

        # Apprentissage
        loss = optimize_model()
        if loss is not None:
            loss_history.append(loss)

        if done:
            break

    rewards_history.append(total_reward)

    # Mise à jour target network
    if episode % TARGET_UPDATE == 0:
        target_net.load_state_dict(policy_net.state_dict())

    # Logging
    if episode % 50 == 0:
        avg_reward = np.mean(rewards_history[-50:])
        print(f"📊 Episode {episode} | Reward moyen (50) : {avg_reward:.2f}")
```

```python

# ==================================================
# Cellule 8 - Visualisation des performances
# ==================================================

plt.figure(figsize=(12,5))

# Courbe brute
plt.plot(rewards_history, label="Reward par épisode")

# Moyenne glissante
window = 50
moving_avg = np.convolve(rewards_history, np.ones(window)/window, mode='valid')
plt.plot(moving_avg, label="Moyenne glissante (50)")

plt.xlabel("Episodes")
plt.ylabel("Reward")
plt.title("Performance DQN - CartPole")
plt.legend()
plt.grid()

plt.show()
```

```python

# ==================================================
# Cellule 7bis - Boucle principale d'entraînement (fournie par mon mentor)
# incluant (DQN avec Gym + PyTorch)
# ==================================================

# ==================================================
# Imports
# ==================================================
import gymnasium as gym
import math
import random
import numpy as np
import matplotlib.pyplot as plt
from collections import deque, namedtuple

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

# ==================================================
# Hyperparamètres
# ==================================================
BATCH_SIZE = 128
GAMMA = 0.99
EPS_START = 0.9
EPS_END = 0.05
EPS_DECAY = 1000
TARGET_UPDATE = 10
LR = 1e-3
NUM_EPISODES = 300
MEMORY_SIZE = 10000

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ==================================================
# Environnement
# ==================================================
env = gym.make("CartPole-v1")

n_actions = env.action_space.n
state_dim = env.observation_space.shape[0]

# ==================================================
# Replay Memory
# ==================================================
Transition = namedtuple('Transition', ('state', 'action', 'reward', 'next_state', 'done'))

class ReplayMemory:
    def __init__(self, capacity):
        self.memory = deque(maxlen=capacity)

    def push(self, *args):
        self.memory.append(Transition(*args))

    def sample(self, batch_size):
        transitions = random.sample(self.memory, batch_size)
        return Transition(*zip(*transitions))

    def __len__(self):
        return len(self.memory)

memory = ReplayMemory(MEMORY_SIZE)

# ==================================================
# Réseau DQN
# ==================================================
class DQN(nn.Module):
    def __init__(self, state_dim, n_actions):
        super(DQN, self).__init__()
        self.fc1 = nn.Linear(state_dim, 128)
        self.fc2 = nn.Linear(128, 128)
        self.fc3 = nn.Linear(128, n_actions)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        return self.fc3(x)

policy_net = DQN(state_dim, n_actions).to(device)
target_net = DQN(state_dim, n_actions).to(device)

target_net.load_state_dict(policy_net.state_dict())
target_net.eval()

optimizer = optim.Adam(policy_net.parameters(), lr=LR)

# ==================================================
# Politique epsilon-greedy
# ==================================================
steps_done = 0

def select_action(state):
    global steps_done
    eps_threshold = EPS_END + (EPS_START - EPS_END) * \
        math.exp(-1. * steps_done / EPS_DECAY)
    steps_done += 1

    if random.random() > eps_threshold:
        with torch.no_grad():
            return policy_net(state).max(1)[1].view(1, 1)
    else:
        return torch.tensor([[random.randrange(n_actions)]], device=device, dtype=torch.long)

# ==================================================
# Optimisation
# ==================================================
def optimize_model():
    if len(memory) < BATCH_SIZE:
        return

    transitions = memory.sample(BATCH_SIZE)
    batch = Transition(*transitions)

    # Masque des états non terminaux
    non_final_mask = torch.tensor(tuple(map(lambda s: s is not None, batch.next_state)),
                                  device=device, dtype=torch.bool)

    non_final_next_states = torch.tensor(
        [s for s in batch.next_state if s is not None],
        device=device, dtype=torch.float32
    ) if any(non_final_mask) else torch.empty((0, state_dim), device=device)

    state_batch = torch.tensor(batch.state, device=device, dtype=torch.float32)
    action_batch = torch.tensor(batch.action, device=device).unsqueeze(1)
    reward_batch = torch.tensor(batch.reward, device=device, dtype=torch.float32)

    # Q(s,a)
    state_action_values = policy_net(state_batch).gather(1, action_batch)

    # max_a Q_target(s', a)
    next_state_values = torch.zeros(BATCH_SIZE, device=device)
    if len(non_final_next_states) > 0:
        next_state_values[non_final_mask] = target_net(non_final_next_states).max(1)[0].detach()

    expected_state_action_values = reward_batch + (GAMMA * next_state_values)

    # Loss
    loss = F.smooth_l1_loss(state_action_values.squeeze(), expected_state_action_values)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# ==================================================
# Boucle principale d'entraînement (fournie par mon mentor)
# ==================================================
episode_rewards = []

for i_episode in range(NUM_EPISODES):
    state, info = env.reset()
    state = torch.tensor(state, dtype=torch.float32, device=device).unsqueeze(0)
    total_reward = 0

    terminated, truncated = False, False

    while not terminated and not truncated:
        action = select_action(state)
        observation, reward, terminated, truncated, _ = env.step(action.item())

        total_reward += reward
        done = terminated or truncated

        # Gestion next_state
        if done:
            next_state = None
        else:
            next_state = observation

        # Stockage
        memory.push(
            state.squeeze(0).cpu().numpy(),
            action.item(),
            reward,
            next_state,
            done
        )

        # Transition
        if next_state is not None:
            state = torch.tensor(next_state, dtype=torch.float32, device=device).unsqueeze(0)
        else:
            state = None

        optimize_model()

        if done:
            break

    episode_rewards.append(total_reward)

    # Update target network
    if i_episode % TARGET_UPDATE == 0:
        target_net.load_state_dict(policy_net.state_dict())

    if (i_episode + 1) % 50 == 0:
        print(f"Épisode {i_episode+1}/{NUM_EPISODES}, Récompense: {total_reward}")

print('Entraînement complet')
env.close()

# ==================================================
# Visualisation
# ==================================================
plt.figure(figsize=(12,6))
plt.plot(episode_rewards)

plt.title('Récompenses par épisode')
plt.xlabel('Épisode')
plt.ylabel('Récompense')

if len(episode_rewards) >= 50:
    moving_avg = np.convolve(episode_rewards, np.ones(50)/50, mode='valid')
    plt.plot(np.arange(len(moving_avg)) + 49, moving_avg, color='red', label='Moyenne mobile (50 épisodes)')

plt.legend()
plt.grid(True)
plt.show()
```

```python

# ==================================================
# Cellule 1 - Setup global + chemins projet
# ==================================================

import os
import json
import time
import gymnasium as gym

from stable_baselines3 import DQN
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

# ==================================================
# 📁 Chemins projet
# ==================================================

BASE_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement"

MISSION_PATH = os.path.join(BASE_PATH, "Mission")

MODEL_BASELINE_PATH = os.path.join(
    MISSION_PATH,
    "models/m01_e01_dqn_lunarlander_baseline.zip"
)

RESULTS_BASELINE_PATH = os.path.join(
    MISSION_PATH,
    "results/m01_e01_baseline_results.json"
)

# 📂 Dossiers sortie Étape 2
MODELS_PATH = os.path.join(MISSION_PATH, "models")
RESULTS_PATH = os.path.join(MISSION_PATH, "results")
LOGS_PATH = os.path.join(MISSION_PATH, "logs_tensorboard")

os.makedirs(MODELS_PATH, exist_ok=True)
os.makedirs(RESULTS_PATH, exist_ok=True)
os.makedirs(LOGS_PATH, exist_ok=True)

print("📁 Dossiers initialisés : ")
print("MISSION_PATH =", MISSION_PATH)
print("MODELS_PATH =", MODELS_PATH)
print("LOGS_PATH =", LOGS_PATH)
print("RESULTS_PATH =", RESULTS_PATH)
```

```python

# ==================================================
# Cellule 3 - Fonction d'expérimentation RL
# ==================================================

def run_experiment(exp_name, hyperparams, total_timesteps=800_000):
    """
    Lance une expérience avec un set d'hyperparamètres donné
    """

    print(f"\n🚀 Lancement expérience : {exp_name}")
    print("Hyperparams :", hyperparams)

    # ==================================================
    # Environnement
    # ==================================================
    env = Monitor(gym.make("LunarLander-v3"))
    eval_env = Monitor(gym.make("LunarLander-v3"))

    # ==================================================
    # Modèle DQN
    # ==================================================
    model = DQN(
        policy="MlpPolicy",
        env=env,
        verbose=0,
        tensorboard_log=LOGS_PATH,
        **hyperparams
    )

    # ==================================================
    # Entraînement
    # ==================================================
    start_time = time.time()

    model.learn(
        total_timesteps=total_timesteps,
        tb_log_name=exp_name
    )

    training_time = time.time() - start_time

    # ==================================================
    # Évaluation
    # ==================================================
    mean_reward, std_reward = evaluate_policy(
        model,
        eval_env,
        n_eval_episodes=100,
        deterministic=True
    )

    print(f"📊 Résultats : mean={mean_reward:.2f}, std={std_reward:.2f}, totaltimesteps=800000")

    # ==================================================
    # Sauvegarde modèle
    # ==================================================
    model_path = os.path.join(MODELS_PATH, f"{exp_name}.zip")
    model.save(model_path)

    # ==================================================
    # Sauvegarde résultats
    # ==================================================
    results = {
        "exp_name": exp_name,
        "hyperparameters": hyperparams,
        "mean_reward": mean_reward,
        "std_reward": std_reward,
        "training_time_sec": training_time
    }

    results_path = os.path.join(RESULTS_PATH, f"{exp_name}.json")

    with open(results_path, "w") as f:
        json.dump(results, f, indent=4)

    print(f"💾 Sauvegardé : {model_path}")
    print(f"📄 Résultats : {results_path}")

    return results
```

```python

# ==================================================
# Cellule 4 - Expérience 1 : Learning rate
# ==================================================

exp1_params = dict(
    # learning_rate=5e-4,
    learning_rate=3e-4,
    gamma=0.99,
    buffer_size=100_000,
    batch_size=64,
    exploration_fraction=0.1,
    exploration_final_eps=0.02,
    target_update_interval=1000
)

exp1_results = run_experiment(
    exp_name="m01_e02_lr_3e4",
    hyperparams=exp1_params
)
```

```python

# ==================================================
# Cellule 5 - Expérience 2 : Gamma
# ==================================================

exp2_params = dict(
    # learning_rate=5e-4,
    learning_rate=3e-4,
    gamma=0.995,  # ↑ vision long terme
    buffer_size=100_000,
    batch_size=64,
    exploration_fraction=0.1,
    exploration_final_eps=0.02,
    target_update_interval=1000
)

exp2_results = run_experiment(
    exp_name="m01_e02_gamma_0995",
    hyperparams=exp2_params
)
```

```python

# ==================================================
# Cellule 2 - Création API FastAPI
# ==================================================

api_code = """
# lunarlander-rl-api/app.py

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field, conlist
import numpy as np
from stable_baselines3 import DQN
import os
import logging

import gymnasium as gym
import imageio
import tempfile


# ==========================
# Initialisation
# ==========================
app = FastAPI(
    title="LunarLander RL API",
    description="API pour prédiction d'action via modèle DQN",
    version="1.0"
)

logging.basicConfig(level=logging.INFO)

# ==========================
# Chargement du modèle
# ==========================
# MODELS_PATH = "m01_e03_best_model.zip"
MODELS_PATH = "models/m01_e03_best_model.zip"

if not os.path.exists(MODELS_PATH):
    raise FileNotFoundError(f"Modèle introuvable: {MODELS_PATH}")

model = DQN.load(MODELS_PATH)

# ==========================
# Schéma entrée utilisateur
# ==========================
class State(BaseModel):
    # liste de 8 floats exactement
    state: conlist(float, min_length=8, max_length=8) = Field(
        ..., description="Vecteur d'état de dimension 8 (LunarLander)"
    )

# ==========================
# Endpoint principal
# ==========================
@app.post("/play")
def play(state: State):
    try:
        # Conversion sécurisée
        obs = np.array(state.state, dtype=np.float32)

        # Vérification explicite (double sécurité)
        if obs.shape != (8,):
            raise HTTPException(status_code=400, detail="State must be length 8")

        # Prédiction
        action, _ = model.predict(obs, deterministic=True)

        return {
            "action": int(action)
        }

    except Exception as e:
        logging.error(f"Erreur dans /play: {str(e)}")
        raise HTTPException(status_code=500, detail=str(e))


# ==========================
# Endpoint GET pour debug navigateur
# ==========================
@app.get("/play")
def play_get():
    return {
        "message": "Utiliser POST avec JSON {'state': [8 valeurs]}"
    }


# ==========================
# Health check
# ==========================
@app.get("/")
def root():
    return {
        "status": "API RL opérationnelle"
    }


# ==========================
# Endpoint info (debug utile)
# ==========================
@app.get("/info")
def info():
    return {
        "model": "DQN LunarLander",
        "input_dim": 8,
        "action_space": 4
    }



# ==========================
# Endpoint full_episode
# ==========================
@app.post("/full_episode")
def full_episode():
    try:
        env = gym.make("LunarLander-v3", render_mode="rgb_array")
        obs, _ = env.reset()

        model_local = model  # ou get_model() si lazy loading

        frames = []
        total_reward = 0

        for _ in range(400):
            action, _ = model_local.predict(obs, deterministic=True)

            obs, reward, done, truncated, _ = env.step(action)
            total_reward += reward

            frame = env.render()
            frames.append(frame)

            if done or truncated:
                break

        env.close()

        # Génération vidéo temporaire
        tmp_file = tempfile.NamedTemporaryFile(suffix=".mp4", delete=False)
        imageio.mimsave(tmp_file.name, frames, fps=30)

        import base64

        with open(tmp_file.name, "rb") as f:
            video_bytes = f.read()

        video_b64 = base64.b64encode(video_bytes).decode()

        return {
            "reward": float(total_reward),
            "video": video_b64
        }

        # return {
        #     "reward": float(total_reward),
        #     "video_path": tmp_file.name
        # }

    except Exception as e:
        logging.error(f"Erreur dans /full_episode: {str(e)}")
        raise HTTPException(status_code=500, detail=str(e))
"""

with open(f"{API_PATH}/app.py", "w") as f:
    f.write(api_code)

app_file = os.path.join(API_PATH, "app.py")
print("✅ Fichier sauvegardé :", app_file)
```

```python

# ==================================================
# Cellule 7 - Interface graphique Streamlit
# ==================================================

import os

gui_code = """
import streamlit as st
import requests
import gymnasium as gym
import numpy as np
import time
import json
import os
import tempfile

# Chemin réel
RESULT_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement/Mission/results/live_metrics.json"

# =========================
# Écriture atomique (safe)
# =========================
def atomic_write_json(path, data):
    os.makedirs(os.path.dirname(path), exist_ok=True)
    with tempfile.NamedTemporaryFile("w", delete=False, dir=os.path.dirname(path)) as tmp:
        json.dump(data, tmp)
        tmp_path = tmp.name
    os.replace(tmp_path, path)

# =========================
# Logging hybride
# =========================
def log_metrics(reward):

    # 1) mémoire temps réel
    st.session_state.rewards_buffer.append(float(reward))

    # 2) flush disque périodique
    step = len(st.session_state.rewards_buffer)

    if step - st.session_state.last_flush_step >= st.session_state.flush_every:

        data = {
            "episode_rewards": list(st.session_state.rewards_buffer)
        }

        atomic_write_json(RESULT_PATH, data)

        st.session_state.last_flush_step = step

# =========================
# GUI principale
# =========================
def run_gui():

    API_URL = "https://remdev-ai-lunarlander-rl-api.hf.space/play"

    st.header("🚀 LunarLander RL Simulation")

    if st.button("Lancer simulation"):

        env = gym.make("LunarLander-v3", render_mode="rgb_array")
        obs, _ = env.reset()

        total_reward = 0
        frame_placeholder = st.empty()

        for step in range(400):

            try:
                response = requests.post(
                    API_URL,
                    json={"state": obs.tolist()},
                    timeout=5
                )

                if response.status_code != 200:
                    st.error(f"Erreur API: {response.text}")
                    break

                action = response.json().get("action", 0)

            except Exception as e:
                st.error(f"Erreur API: {e}")
                break

            obs, reward, done, truncated, _ = env.step(action)
            total_reward += reward

            frame = env.render()
            frame_placeholder.image(frame, channels="RGB")

            time.sleep(0.03)

            if done or truncated:
                break

        env.close()

        # 🔥 POINT CRITIQUE : log après épisode
        log_metrics(total_reward)

        st.success(f"Reward total: {total_reward:.2f}")
        st.info(f"Buffer size: {len(st.session_state.rewards_buffer)}")
"""

# Création dossier si nécessaire
os.makedirs(GUI_PATH, exist_ok=True)

# Écriture fichier
with open(f"{GUI_PATH}/app.py", "w") as f:
    f.write(gui_code)

GUI_APP = os.path.join(GUI_PATH, "app.py")
print("✅ GUI créée :", GUI_APP)
```

```python

# ==================================================
# Cellule 8 - Dashboard analytique (corrigée HF)
# ==================================================

import os

dashboard_code = """
import streamlit as st
import pandas as pd
import json
import os
from pathlib import Path

# Chemin ABSOLU (robuste Colab)
# RESULT_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_11_entrainez_votre_agent_RL_apprentissage_par_renforcement/Mission/results/m01_e02_gamma_0995.json"

BASE_DIR = Path(__file__).resolve().parent  # racine script
RESULT_PATH = BASE_DIR / "results" / "m01_e02_gamma_0995.json"

def load_from_disk():
    if not os.path.exists(RESULT_PATH):
        return []

    try:
        with open(RESULT_PATH, "r") as f:
            data = json.load(f)
        return data.get("episode_rewards", [])
    except Exception as e:
        st.warning(f"Erreur lecture JSON: {e}")
        return []


def run_dashboard():

    st.header("📊 Dashboard RL (temps réel)")

    # =========================
    # CONTROLES UI
    # =========================
    col1, col2, col3 = st.columns(3)

    with col1:
        auto_refresh = st.toggle("Auto-refresh", value=st.session_state.auto_refresh)

    with col2:
        if st.button("🔄 Refresh"):
            st.rerun()

    with col3:
        max_points = st.number_input("Fenêtre", 100, 5000, 1000, step=100)

    st.session_state.auto_refresh = auto_refresh

    # =========================
    # SOURCE DES DONNÉES
    # =========================
    if "rewards_buffer" in st.session_state and len(st.session_state.rewards_buffer) > 0:
        rewards = list(st.session_state.rewards_buffer)
        source = "mémoire (temps réel)"
    else:
        rewards = load_from_disk()
        source = "disque (fallback)"

    if len(rewards) == 0:
        st.warning("Aucune donnée disponible.")
        return

    # =========================
    # FENETRAGE
    # =========================
    rewards = rewards[-int(max_points):]

    df = pd.DataFrame(rewards, columns=["reward"])

    st.caption(f"Source: {source} | points: {len(df)}")

    # =========================
    # STATISTIQUES
    # =========================
    st.subheader("📈 Statistiques")
    st.dataframe(df.describe(), width="stretch")

    # =========================
    # COURBE
    # =========================
    st.subheader("📉 Convergence")
    st.line_chart(df, width="stretch")

    # =========================
    # DISTRIBUTION
    # =========================
    st.subheader("📊 Distribution")
    st.bar_chart(df, width="stretch")

    # =========================
    # AUTO REFRESH DOUX
    # =========================
    if st.session_state.auto_refresh:
        st.caption("Auto-refresh actif (interaction requise)")
"""

# Création dossier si nécessaire
os.makedirs(DASHBOARD_PATH, exist_ok=True)

# Écriture fichier
with open(f"{DASHBOARD_PATH}/app.py", "w") as f:
    f.write(dashboard_code)

Dashboard_APP = os.path.join(DASHBOARD_PATH, "app.py")
print("✅ Dashboard créé :", Dashboard_APP)
```

```python

# ==================================================
# Cellule 9 - (version 1) Génération vidéo (HF compatible)
# ==================================================

import os

video_code = """
import gymnasium as gym
import imageio
import requests
import os
import numpy as np

API_URL = "https://remdev-ai-lunarlander-rl-api.hf.space/play"

OUTPUT_DIR = "outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
OUTPUT_PATH = os.path.join(OUTPUT_DIR, "demo.mp4")

MAX_STEPS = 400
FPS = 30


def pad_frame(frame):
    h, w, _ = frame.shape
    new_w = (w + 15) // 16 * 16
    new_h = (h + 15) // 16 * 16

    padded = np.zeros((new_h, new_w, 3), dtype=frame.dtype)
    padded[:h, :w] = frame
    return padded


def get_action(obs, env):
    try:
        r = requests.post(
            API_URL,
            json={"state": obs.tolist()},
            timeout=5
        )
        r.raise_for_status()
        return int(r.json().get("action", 0))

    except Exception as e:
        print(f"[API ERROR] {e}")
        return env.action_space.sample()


def generate_video():

    env = gym.make("LunarLander-v3", render_mode="rgb_array")
    obs, _ = env.reset()

    total_reward = 0

    with imageio.get_writer(OUTPUT_PATH, fps=FPS) as writer:

        for step in range(MAX_STEPS):

            action = get_action(obs, env)

            obs, reward, done, truncated, _ = env.step(action)
            total_reward += reward

            frame = env.render()
            frame = pad_frame(frame)

            writer.append_data(frame)

            if done or truncated:
                break

    env.close()

    return OUTPUT_PATH, total_reward
"""

# Création dossier
VIDEO_PATH = os.path.join(MISSION_PATH, "videos")
os.makedirs(VIDEO_PATH, exist_ok=True)

# Écriture fichier
with open(f"{VIDEO_PATH}/generator.py", "w") as f:
    f.write(video_code)

# print("✅ Module vidéo créé : videos/generator.py")
video_generator = os.path.join(VIDEO_PATH, "generator.py")
print("✅ Module vidéo créé :", video_generator)
```

```python

# ==================================================
# Cellule 9bis - (version 2) Génération vidéo démo (localement)
# ==================================================

import gymnasium as gym
from gymnasium.wrappers import RecordVideo
import requests
import os
import time
import glob
import shutil
import imageio


# ==========================
# Configuration
# ==========================
API_URL = "https://remdev-ai-lunarlander-rl-api.hf.space/play"

VIDEO_PATH = os.path.join(MISSION_PATH, "videos")
os.makedirs(VIDEO_PATH, exist_ok=True)

MAX_ATTEMPTS = 10
TARGET_MIN_REWARD = 200
MAX_STEPS = 1000 #1000
TIMEOUT = 15
SEED = None  # ⚠️ important : pas de seed fixe

FPS = 30  # 50
MIN_FRAMES = 20 * FPS
MAX_FRAMES = 30 * FPS

# ==========================
# API robuste
# ==========================
def get_action(obs, env):
    try:
        r = requests.post(API_URL, json={"state": obs.tolist()}, timeout=TIMEOUT)
        if r.status_code == 200:
            return int(r.json().get("action", 0))
    except:
        pass
    return env.action_space.sample()

# ==========================
# Recherche épisode valide
# ==========================
best_reward = -float("inf")
best_video = None

for attempt in range(MAX_ATTEMPTS):

    run_name = f"m01_e03_demo_lunarlander_Episode-{attempt+1}_{int(time.time())}"

    env = gym.make("LunarLander-v3", render_mode="rgb_array")

    env = RecordVideo(
        env,
        video_folder=VIDEO_PATH,
        name_prefix=run_name,
        episode_trigger=lambda x: True
    )

    obs, _ = env.reset(seed=SEED)

    total_reward = 0
    frame_count = 0

    for step in range(MAX_STEPS):

        action = get_action(obs, env)

        obs, reward, done, truncated, _ = env.step(action)

        total_reward += reward
        frame_count += 1

        if done or truncated:
            break

        if frame_count >= MAX_FRAMES:
            break

    env.close()

    # récupérer le fichier généré
    video_files = glob.glob(os.path.join(VIDEO_PATH, f"{run_name}*.mp4"))

    real_fps = None
    duration = None

    if video_files:
        original_file = video_files[0]
        clean_file = os.path.join(VIDEO_PATH, f"{run_name}.mp4")

        shutil.move(original_file, clean_file)

        # 🔍 Lecture du FPS réel de la vidéo
        try:
            reader = imageio.get_reader(clean_file)
            meta = reader.get_meta_data()
            real_fps = meta.get("fps", None)
            reader.close()
        except Exception as e:
            print("⚠️ Impossible de lire le FPS réel:", e)

        # 🧮 Calcul durée réelle
        if real_fps:
            duration = frame_count / real_fps
        else:
            duration = frame_count / FPS  # fallback
    else:
        duration = frame_count / FPS  # fallback si vidéo absente

    if real_fps:
        min_frames_real = int(20 * real_fps)
        max_frames_real = int(30 * real_fps)
    else:
        min_frames_real = MIN_FRAMES
        max_frames_real = MAX_FRAMES

    fps_info = f"{real_fps:.1f}" if real_fps else "N/A"
    print(f"🎯 Episode {attempt+1} | Reward: {total_reward:.2f} | Durée: {duration:.2f}s | FPS réel: {fps_info}")

    # 🎯 Critère strict
    # if (
    #     total_reward >= TARGET_MIN_REWARD and
    #     min_frames_real <= frame_count <= max_frames_real
    # ):
    if (
        total_reward >= TARGET_MIN_REWARD and
        20 <= duration <= 30
    ):
        print("✅ Épisode valide trouvé (performance + durée)")
        best_reward = total_reward
        best_video = os.path.join(VIDEO_PATH, run_name)
        break

    # fallback (meilleur même si non valide)
    if total_reward > best_reward:
        best_reward = total_reward
        best_video = os.path.join(VIDEO_PATH, run_name)

# ==========================
# Résultat final
# ==========================
print("\n🎥 Résultat final")
print(f"Reward: {best_reward:.2f}")
print("Vidéo:", best_video)
```
