````python

# ============================================================
# Cellule 6 — 🕷️ SPIDER MULTI-SOURCES (V2 ÉQUILIBRÉ)
# ============================================================

spider_code = """
import scrapy
from urllib.parse import urlparse
from news_scraper.items import NewsItem
import re


class NewsSpider(scrapy.Spider):
    name = "news_multi"

    allowed_domains = [
        "reuters.com", "apnews.com", "lefigaro.fr",
        "cnn.com", "abcnews.com", "nbcnews.com"
    ]

    start_urls = [
        "https://www.reuters.com/world/",
        "https://apnews.com/hub/world-news",
        "https://www.lefigaro.fr/international/",
        "https://edition.cnn.com/world",
        "https://abcnews.go.com/International",
        "https://www.nbcnews.com/world"
    ]

    # ✅ Patterns élargis (équilibre recall / précision)
    ARTICLE_PATTERNS = [
        r"/\\d{4}/\\d{2}/\\d{2}/",
        r"/article/",
        r"/story/",
        r"/news/",
    ]

    def is_article_url(self, url):
        return any(re.search(p, url) for p in self.ARTICLE_PATTERNS)

    def is_valid_domain(self, url):
        netloc = urlparse(url).netloc
        return any(domain in netloc for domain in self.allowed_domains)

    def parse(self, response):

        links = response.css("a::attr(href)").getall()

        for link in links:
            if not link:
                continue

            link = response.urljoin(link)

            # ✅ FIX domaine (support sous-domaines)
            if not self.is_valid_domain(link):
                continue

            if self.is_article_url(link):
                yield scrapy.Request(
                    link,
                    callback=self.parse_article,
                    errback=self.errback,
                    headers={
                        "Referer": response.url,
                        "User-Agent": self.settings.get("USER_AGENT"),
                    }
                )
            else:
                # ✅ crawl élargi mais contrôlé
                if any(x in link for x in [
                    "world", "international", "news", "latest"
                ]):
                    yield scrapy.Request(link, callback=self.parse)

    def errback(self, failure):
        self.logger.warning(f"Request failed: {failure.request.url}")

    def parse_article(self, response):

        item = NewsItem()

        parsed_url = urlparse(response.url)

        domain_full = parsed_url.netloc
        domain = domain_full.replace("www.", "")

        item["domain_full"] = domain_full
        item["source"] = ".".join(domain.split(".")[-2:])
        item["site"] = domain.split(".")[0]
        item["url"] = response.url

        # ✅ titre robuste
        title = (
            response.css("h1::text").get() or
            response.css("meta[property='og:title']::attr(content)").get()
        )
        item["title"] = title.strip() if title else ""

        # ✅ extraction plus large
        paragraphs = (
            response.css("article p::text").getall() or
            response.css("div p::text").getall() or
            response.css("p::text").getall()
        )

        cleaned = []
        for p in paragraphs:
            p = p.strip()

            if len(p) < 30:
                continue

            if any(x in p.lower() for x in [
                "cookie", "privacy", "advertisement", "subscribe", "newsletter"
            ]):
                continue

            cleaned.append(p)

        text = " ".join(cleaned)

        if len(text) < 60:
            return

        item["text"] = text
        item["content_length"] = len(text)

        item["published"] = (
            response.css("time::attr(datetime)").get() or
            response.css("meta[property='article:published_time']::attr(content)").get()
        )

        image = (
            response.css("meta[property='og:image']::attr(content)").get() or
            response.css("img::attr(src)").get()
        )

        item["image_urls"] = [response.urljoin(image)] if image else []

        yield item
"""
with open("news_scraper/spiders/news_spider.py", "w") as f:
    f.write(spider_code)
````

````python

# ============================================================
# Cellule 7 — 🖼️ PIPELINE IMAGES (ROBUSTE)
# ============================================================

pipeline_code = """
from scrapy.pipelines.images import ImagesPipeline
import scrapy


class ImagePipeline(ImagesPipeline):

    def get_media_requests(self, item, info):
        for url in item.get("image_urls", []):
            if url:
                yield scrapy.Request(url, dont_filter=True)

    def item_completed(self, results, item, info):

        images = []

        for ok, data in results:
            if ok and isinstance(data, dict):
                images.append(data.get("path"))

        item["images"] = images
        return item

    def media_failed(self, failure, request, info):
        return None
"""
with open("news_scraper/pipelines.py", "w") as f:
    f.write(pipeline_code)
````

````python

# ====================================================================
# Cellule 6 — 📊 Harmonisation + Fusion de Kaggle et Dataset Scraped
# ====================================================================

import os
import json
import pandas as pd

BASE_PATH = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission"
BASE_PATH_E03 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-03"
BASE_PATH_E02 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-02"
BASE_PATH_E01 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-01"

PATH_JSON = f"{BASE_PATH_E02}/outputs/e02_scraped_data.json"
PATH_CSV = f"{BASE_PATH_E01}/raw_data/Fake_news_content_detection.csv"

# -------------------------------
# 1. Chargement SCRAPED
# -------------------------------
with open(PATH_JSON, "r", encoding="utf-8") as f:
    scraped = json.load(f)

df_scraped = pd.DataFrame(scraped)

# Normalisation
df_scraped["label"] = None

# -------------------------------
# 2. Chargement Kaggle
# -------------------------------
# fake_path = BASE_PATH_E01 + "/raw_data/Fake_news_content_detection.csv"
df_fake = pd.read_csv(PATH_CSV)

# -------------------------------
# 3. Reconstruction du label
# -------------------------------

label_columns = [
    "Barely-True",
    "False",
    "Half-True",
    "Mostly-True",
    "Not-Known",
    "True"
]

# Convertir one-hot → label texte
df_fake["label_raw"] = df_fake[label_columns].idxmax(axis=1)

# -------------------------------
# 4. Mapping vers label binaire (recommandé ML)
# -------------------------------

def map_label(x):
    if x in ["False", "Barely-True"]:
        return 0   # fake
    elif x in ["Mostly-True", "True"]:
        return 1   # real
    else:
        return None  # Half-True / Not-Known → ambigu

df_fake["label"] = df_fake["label_raw"].apply(map_label)

# -------------------------------
# 5. Nettoyage
# -------------------------------
df_fake = df_fake[df_fake["label"].notnull()]

# -------------------------------
# 6. Harmonisation colonnes
# -------------------------------
df_fake = df_fake.rename(columns={
    "Text": "text",
    "Text_Tag": "source"
})

df_fake["source"] = "kaggle_fake_news"
df_fake["title"] = None
df_fake["url"] = None
df_fake["published"] = None
df_fake["image_urls"] = None
df_fake["images"] = None

# Sélection finale
df_fake = df_fake[[
    "source", "title", "text", "url",
    "published", "image_urls", "images", "label"
]]

# -------------------------------
# 7. Harmonisation SCRAPED
# -------------------------------
for col in df_fake.columns:
    if col not in df_scraped.columns:
        df_scraped[col] = None

df_scraped = df_scraped[df_fake.columns]

# -------------------------------
# 8. Fusion (VERSION MULTIMODALE)
# -------------------------------

df_scraped["label"] = None  # non supervisé

df_final = pd.concat([df_scraped, df_fake], ignore_index=True)

# Ajouter flag supervision
df_final["is_labeled"] = df_final["label"].notna().astype(int)

print()

# -------------------------------
# 9. Vérification
# -------------------------------

print("Dataset final :", df_final.shape)

# Mapping inverse
label_map_text = {
    0.0: "FAKE",
    1.0: "REAL"
}

# Conversion pour affichage
df_final["label_text"] = df_final["label"].map(label_map_text)

# Remplacer NaN → UNLABELED
df_final["label_text"] = df_final["label_text"].fillna("UNLABELED")

print("\nDistribution labels (texte) :")
print(df_final["label_text"].value_counts())

print("\nLabeled vs Unlabeled :")
print(df_final["is_labeled"].value_counts())


# -------------------------------
# 10. Sauvegarde
# -------------------------------
OUTPUT_DATASET_PATH = BASE_PATH_E03 + "/outputs/e03_final_dataset.csv"
OUTPUT_JSONL_PATH = BASE_PATH_E03 + "/outputs/e03_final_dataset.jsonl"

# Création dossier si nécessaire
os.makedirs(BASE_PATH + "/outputs", exist_ok=True)

# CSV
df_final.to_csv(OUTPUT_DATASET_PATH, index=False, encoding="utf-8")

# JSONL (recommandé ML)
with open(OUTPUT_JSONL_PATH, "w", encoding="utf-8") as f:
    for record in df_final.to_dict(orient="records"):
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print()

print("✅ Dataset sauvegardé :")
print("CSV :", OUTPUT_DATASET_PATH)
print("JSONL :", OUTPUT_JSONL_PATH)
````

````python

# ============================================================
# Cellule 7 — ⚙️ PIPELINE DE TRANSFORMATION MODULAIRE
# ============================================================

import os
import pandas as pd
import logging
import re
import json

# -------------------------------
# 1. CONFIGURATION
# -------------------------------
BASE_PATH_E03 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-03"
INPUT_PATH = BASE_PATH_E03 + "/outputs/e03_final_dataset.csv"

OUTPUT_CLEAN_CSV = BASE_PATH_E03 + "/outputs/e03_final_pipeline_dataset.csv"
OUTPUT_CLEAN_JSONL = BASE_PATH_E03 + "/outputs/e03_final_pipeline_dataset.jsonl"

LOG_PATH = BASE_PATH_E03 + "/outputs/e03_pipeline_transformation.log"

os.makedirs(BASE_PATH_E03 + "/outputs", exist_ok=True)

# -------------------------------
# 2. LOGGING
# -------------------------------
logging.basicConfig(
    filename=LOG_PATH,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("🚀 Démarrage du pipeline de transformation")

# -------------------------------
# 3. FONCTIONS MODULAIRES
# -------------------------------
def nettoie_texte(text):
    """Nettoyage NLP de base"""
    if pd.isna(text):
        return None

    text = str(text)

    # Suppression HTML
    text = re.sub(r"<.*?>", "", text)

    # Suppression caractères spéciaux
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    # Normalisation espaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


def valide_image(image_urls):
    """Validation basique des URLs images"""
    if pd.isna(image_urls):
        return False

    if isinstance(image_urls, str):
        return image_urls.startswith("http")

    return False


def enrich_features(df):
    """Feature engineering utile pour ML"""

    logging.info("🧠 Feature engineering")

    df["text_length"] = df["text"].apply(lambda x: len(x) if pd.notna(x) else 0)

    df["has_image"] = df["image_urls"].apply(valide_image).astype(int)

    df["nb_images"] = df["image_urls"].apply(
        lambda x: len(x.split(",")) if isinstance(x, str) else 0
    )

    return df


def verification_dataset(df):
    """Contrôles qualité"""

    logging.info("🔍 Vérification dataset")

    assert "text" in df.columns, "Colonne text manquante"
    assert "label" in df.columns, "Colonne label manquante"

    # Vérification texte vide
    empty_text = df["text"].isna().sum()
    logging.info(f"Textes vides : {empty_text}")

    # Vérification cohérence image
    mismatch = df[(df["has_image"] == 1) & (df["nb_images"] == 0)].shape[0]
    logging.info(f"Incohérences image : {mismatch}")

    return df


# -------------------------------
# 4. PIPELINE
# -------------------------------
def pipeline_transformation(path):

    logging.info("📥 Lecture du dataset")
    df = pd.read_csv(path)

    # ---------------------------
    # Nettoyage
    # ---------------------------
    logging.info("🧹 Nettoyage des données")

    df["text"] = df["text"].apply(nettoie_texte)
    df["title"] = df["title"].apply(nettoie_texte)

    # ---------------------------
    # Feature engineering
    # ---------------------------
    df = enrich_features(df)

    # ---------------------------
    # Vérification
    # ---------------------------
    df = verification_dataset(df)

    # ---------------------------
    # Export
    # ---------------------------
    logging.info("💾 Export des données")

    df.to_csv(OUTPUT_CLEAN_CSV, index=False, encoding="utf-8")

    with open(OUTPUT_CLEAN_JSONL, "w", encoding="utf-8") as f:
        for record in df.to_dict(orient="records"):
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

    logging.info("✅ Pipeline terminé")

    return df


# -------------------------------
# 5. EXECUTION
# -------------------------------
df_clean = pipeline_transformation(INPUT_PATH)

print("✅ Dataset transformé :", df_clean.shape)
print("📁 CSV :", OUTPUT_CLEAN_CSV)
print("📁 JSONL :", OUTPUT_CLEAN_JSONL)
print("📄 LOG :", LOG_PATH)
````

````python

# ============================================================
# Cellule 8 — 🧠 EXPORT PNG MERMAID + MARGES + TITRE
# ============================================================

import base64
import requests
import os
import time
from PIL import Image, ImageOps, ImageDraw, ImageFont

# -------------------------------
# CONFIG
# -------------------------------
BASE_PATH_E03 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-03"
OUTPUT_PATH = BASE_PATH_E03 + "/outputs/e03_mermaid_diagram.png"

os.makedirs(BASE_PATH_E03 + "/outputs", exist_ok=True)

TITLE_TEXT = "Schéma de données : Modèle conceptuel"
TITLE_FONT_SIZE = 80
TOP_PADDING = 220
MARGIN = 80

MAX_RETRIES = 5
TIMEOUT = 10

# -------------------------------
# MERMAID CODE
# -------------------------------
mermaid_code = (
"erDiagram\n"
"\n"
"ARTICLE {\n"
"  string titre\n"
"  string contenu_textuel\n"
"  string source\n"
"  string url\n"
"  date date_publication\n"
"}\n"
"\n"
"IMAGE {\n"
"  string url_image\n"
"  string description_image\n"
"}\n"
"\n"
"CLASSIFICATION {\n"
"  string categorie_information\n"
"  float niveau_confiance\n"
"  string statut_annotation\n"
"}\n"
"\n"
"DONNEE_PROVENANCE {\n"
"  string origine_donnee\n"
"  string etape_traitement\n"
"  date date_collecte\n"
"}\n"
"\n"
"ARTICLE ||--o{ IMAGE : illustre\n"
"ARTICLE ||--o{ CLASSIFICATION : est_evalue\n"
"ARTICLE ||--o{ DONNEE_PROVENANCE : est_trace\n"
)

# -------------------------------
# ENCODAGE
# -------------------------------
encoded = base64.b64encode(mermaid_code.encode("utf-8")).decode("utf-8")

urls = [
    f"https://mermaid.ink/img/{encoded}?width=1600&scale=3&bgColor=white",
    f"https://mermaid.live/img/{encoded}?width=1600&scale=3&bgColor=white"
]

# -------------------------------
# DOWNLOAD AVEC RETRY
# -------------------------------
response = None

for url in urls:
    print(f"🌐 Tentative avec : {url.split('/')[2]}")

    for attempt in range(MAX_RETRIES):
        try:
            response = requests.get(url, timeout=TIMEOUT)

            if response.status_code == 200:
                print("✅ Succès téléchargement")
                break
            else:
                print(f"⚠️ Tentative {attempt+1} - Status {response.status_code}")

        except requests.exceptions.RequestException as e:
            print(f"⚠️ Erreur réseau : {e}")

        time.sleep(2 ** attempt)

    if response and response.status_code == 200:
        break

if not response or response.status_code != 200:
    raise Exception("❌ Échec complet génération Mermaid")

# -------------------------------
# SAVE IMAGE
# -------------------------------
with open(OUTPUT_PATH, "wb") as f:
    f.write(response.content)

print("✅ PNG Mermaid généré")

# -------------------------------
# AJOUT MARGES
# -------------------------------
img = Image.open(OUTPUT_PATH)
img = ImageOps.expand(img, border=MARGIN, fill='white')

# -------------------------------
# AJOUT TITRE
# -------------------------------
new_img = Image.new("RGB", (img.width, img.height + TOP_PADDING), "white")
new_img.paste(img, (0, TOP_PADDING))

draw = ImageDraw.Draw(new_img)

# -------------------------------
# CHARGEMENT POLICE (MULTI-CHEMINS)
# -------------------------------
font_paths = [
    "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf",
    "/usr/share/fonts/TTF/DejaVuSans-Bold.ttf",
    "/usr/share/fonts/truetype/liberation/LiberationSans-Bold.ttf"
]

font = None

for path in font_paths:
    if os.path.exists(path):
        font = ImageFont.truetype(path, TITLE_FONT_SIZE)
        print(f"✅ Police chargée : {path}")
        break

# fallback propre
if font is None:
    print("⚠️ Aucune police système trouvée → fallback PIL")
    font = ImageFont.load_default()
    TITLE_FONT_SIZE = 40  # éviter texte trop petit

# -------------------------------
# AJUSTEMENT DYNAMIQUE
# -------------------------------
max_width = int(new_img.width * 0.9)

while True:
    bbox = draw.textbbox((0, 0), TITLE_TEXT, font=font)
    text_width = bbox[2] - bbox[0]

    if text_width <= max_width or TITLE_FONT_SIZE <= 20:
        break

    TITLE_FONT_SIZE -= 2

    if font_paths:
        for path in font_paths:
            if os.path.exists(path):
                font = ImageFont.truetype(path, TITLE_FONT_SIZE)
                break

# Position finale
bbox = draw.textbbox((0, 0), TITLE_TEXT, font=font)
text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]

x_center = (new_img.width - text_width) // 2
y_position = (TOP_PADDING - text_height) // 2

draw.text((x_center, y_position), TITLE_TEXT, fill="black", font=font)

# -------------------------------
# SAVE FINAL
# -------------------------------
new_img.save(OUTPUT_PATH, dpi=(300, 300))

print("✅ Image finale générée :", OUTPUT_PATH)
````

````python

# ============================================
# Cellule 4 — 📥 SAUVEGARDE DU DAG (CORRIGÉ FINAL)
# ============================================

dag_code = """
from airflow import DAG
from airflow.operators.python import PythonOperator
from datetime import datetime
import pandas as pd
import sqlite3
import logging
import os
import json
from urllib.parse import urlparse

# ============================================
# CONFIGURATION
# ============================================

BASE_PATH = "/home/rdev-xxx/airflow"
CSV_PATH = f"{BASE_PATH}/data/e03_final_pipeline_dataset.csv"
DB_PATH = f"{BASE_PATH}/data/e04_etl_pipeline_v01.db"


# ============================================
# ETAPE 1 : EXTRACTION
# ============================================

def extract_data():
    logging.info("Lecture du fichier CSV...")

    if not os.path.exists(CSV_PATH):
        raise FileNotFoundError(f"Fichier introuvable : {CSV_PATH}")

    df = pd.read_csv(CSV_PATH)

    logging.info(f"Shape extrait : {df.shape}")
    logging.info(f"Colonnes : {list(df.columns)}")

    if "label_text" in df.columns:
        logging.info(f"Distribution labels :\\n{df['label_text'].value_counts(dropna=False)}")

    df.to_csv("/tmp/extracted.csv", index=False)


# ============================================
# ETAPE 2 : TRANSFORMATION (CORRIGÉE)
# ============================================

def transform_data():
    logging.info("Transformation avancée...")

    df = pd.read_csv("/tmp/extracted.csv")

    logging.info(f"Shape AVANT transformation : {df.shape}")

    # ========================================
    # 1. Déduplication intelligente (FIX)
    # ========================================
    if "url" in df.columns:
        df_with_url = df[df["url"].notna()].drop_duplicates(subset=["url"])
        df_without_url = df[df["url"].isna()]

        df = pd.concat([df_with_url, df_without_url], ignore_index=True)

        logging.info(f"Après dédup URL : {df.shape}")
    else:
        df = df.drop_duplicates()
        logging.info(f"Après dédup globale : {df.shape}")

    # ========================================
    # 2. Nettoyage texte
    # ========================================
    def clean_text(text):
        if not isinstance(text, str):
            return ""
        if text.startswith("Copyright"):
            return ""
        return text.strip()

    df["text"] = df.get("text", "").apply(clean_text)

    # ========================================
    # 3. Dates
    # ========================================
    df["published"] = pd.to_datetime(df.get("published"), errors="coerce")

    # ========================================
    # 4. Images (robuste)
    # ========================================
    def ensure_list(x):
        if isinstance(x, list):
            return x
        if isinstance(x, str):
            try:
                return json.loads(x)
            except:
                try:
                    return eval(x)
                except:
                    return []
        return []

    df["images"] = df.get("images", []).apply(ensure_list)

    # ========================================
    # 5. Feature engineering
    # ========================================
    df["nb_images"] = df["images"].apply(lambda x: len(x) if isinstance(x, list) else 0)
    df["has_image"] = df["nb_images"].apply(lambda x: 1 if x > 0 else 0)
    df["text_length"] = df["text"].apply(lambda x: len(x) if isinstance(x, str) else 0)

    # ========================================
    # 6. Domain extraction
    # ========================================
    def extract_domain(url):
        try:
            return urlparse(url).netloc
        except:
            return None

    df["domain_full"] = df.get("url").apply(extract_domain)

    # ========================================
    # 7. Labels
    # ========================================
    label_mapping = {"fake": 0, "real": 1, "unlabeled": -1, "unknown": -1}

    df["label_text"] = df.get("label_text", "unlabeled").astype(str).str.lower()
    df["label"] = df["label_text"].map(label_mapping).fillna(-1)
    df["is_labeled"] = (df["label"] != -1).astype(int)

    # ========================================
    # 8. Metadata
    # ========================================
    df["processed_at"] = datetime.utcnow()

    # ========================================
    # 9. JSON SQLite
    # ========================================
    df["images_json"] = df["images"].apply(
        lambda x: json.dumps(x) if isinstance(x, list) else "[]"
    )

    logging.info(f"Shape FINAL transformé : {df.shape}")

    df.to_csv("/tmp/transformed.csv", index=False)


# ============================================
# ETAPE 3 : LOAD
# ============================================

def load_data():
    logging.info("Chargement dans SQLite...")

    conn = sqlite3.connect(DB_PATH)
    cursor = conn.cursor()

    cursor.execute("DROP TABLE IF EXISTS multimodal_data")

    cursor.execute(\"\"\"
    CREATE TABLE multimodal_data (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        url TEXT,
        domain_full TEXT,
        source TEXT,
        title TEXT,
        text TEXT,
        published TEXT,
        images_json TEXT,
        nb_images INTEGER,
        has_image INTEGER,
        text_length INTEGER,
        label INTEGER,
        label_text TEXT,
        is_labeled INTEGER,
        processed_at TEXT
    );
    \"\"\")

    df = pd.read_csv("/tmp/transformed.csv")

    cols = [
        "url", "domain_full", "source",
        "title", "text", "published",
        "images_json",
        "nb_images", "has_image", "text_length",
        "label", "label_text", "is_labeled",
        "processed_at"
    ]

    for col in cols:
        if col not in df.columns:
            df[col] = None

    df = df.where(pd.notnull(df), None)

    rows = df[cols].values.tolist()

    cursor.executemany(
        \"\"\"
        INSERT INTO multimodal_data (
            url, domain_full, source,
            title, text, published,
            images_json,
            nb_images, has_image, text_length,
            label, label_text, is_labeled,
            processed_at
        ) VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
        \"\"\",
        rows
    )

    conn.commit()
    conn.close()

    logging.info(f"Lignes chargées : {len(df)}")
    logging.info("Chargement terminé.")


# ============================================
# DAG
# ============================================

default_args = {
    "owner": "airflow",
    "start_date": datetime(2024, 1, 1),
    "retries": 1
}

with DAG(
    dag_id="e04_etl_multimodal_pipeline_v01",
    default_args=default_args,
    schedule="@daily",
    catchup=False
) as dag:

    extract = PythonOperator(
        task_id="extract",
        python_callable=extract_data
    )

    transform = PythonOperator(
        task_id="transform",
        python_callable=transform_data
    )

    load = PythonOperator(
        task_id="load",
        python_callable=load_data
    )

    extract >> transform >> load
"""

# Sauvegarde du DAG
with open(f"{DAGS_FOLDER}/e04_etl_pipeline_v01.py", "w") as f:
    f.write(dag_code)

print("DAG corrigé sauvegardé :", f"{DAGS_FOLDER}/e04_etl_pipeline_v01.py")
````

````python

# ================================
# Cellule 5b — 🔍 INSPECTION SQLITE
# ================================

import os
import sqlite3
import pandas as pd

BASE_PATH_E04 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-04/airflow/dags"
DB_PATH_E04 = f"{BASE_PATH_E04}/e04_etl_pipeline_v01.db"

print(f"📍 Chemin : {DB_PATH_E04}")
print(f"📦 Taille fichier : {round(os.path.getsize(DB_PATH_E04) / (1024**2), 2)} MB")

conn = sqlite3.connect(DB_PATH_E04)

# ================================
# 1️⃣ Volume global
# ================================
print("\n📈 Volume total :")
print(pd.read_sql("SELECT COUNT(*) as total FROM multimodal_data;", conn))

# ================================
# 2️⃣ Distribution labels (FIABLE)
# ================================
print("\n🎯 Distribution réelle :")

df_labels = pd.read_sql("""
SELECT
    label_text,
    COUNT(*) as count
FROM multimodal_data
GROUP BY label_text
ORDER BY count DESC;
""", conn)

print(df_labels)

# ================================
# 3️⃣ Vérification cohérence Kaggle
# ================================
print("\n🧪 Cohérence Kaggle :")

df_kaggle_check = pd.read_sql("""
SELECT
    COUNT(*) as total_kaggle,
    SUM(CASE WHEN is_labeled = 1 THEN 1 ELSE 0 END) as labeled_ok,
    SUM(CASE WHEN is_labeled = 0 THEN 1 ELSE 0 END) as labeled_missing
FROM multimodal_data
WHERE source = 'kaggle_fake_news';
""", conn)

print(df_kaggle_check)

# ================================
# 4️⃣ Vérification incohérences RÉELLES
# ================================
print("\n🚨 Incohérences critiques (label contradictoire) :")

df_inconsistency = pd.read_sql("""
SELECT COUNT(*) as inconsistent_rows
FROM multimodal_data
WHERE
    (is_labeled = 1 AND label = -1)
    OR
    (is_labeled = 0 AND label != -1);
""", conn)

print(df_inconsistency)

# ================================
# 5️⃣ Scrap riche NON labellisé (normal)
# ================================
print("\n📄 Articles scrap riches non labellisés :")

df_scrap_rich = pd.read_sql("""
SELECT source, COUNT(*) as count
FROM multimodal_data
WHERE
    is_labeled = 0
    AND text_length > 1000
GROUP BY source
ORDER BY count DESC;
""", conn)

print(df_scrap_rich)

# ================================
# 6️⃣ Qualité globale
# ================================
print("\n🧪 Qualité données :")

df_quality = pd.read_sql("""
SELECT
    COUNT(*) as total,
    SUM(CASE WHEN text IS NULL OR text = '' THEN 1 ELSE 0 END) as empty_text,
    SUM(CASE WHEN nb_images = 0 THEN 1 ELSE 0 END) as no_images
FROM multimodal_data;
""", conn)

print(df_quality)

# ================================
# FIN
# ================================
conn.close()
````

````python

# ================================
# Cellule 6 — 📊 KPI ETL + DASHBOARD + MONITORING
# ================================

import time
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3
from datetime import datetime
import os

start_time = time.time()

# ================================
# 0️⃣ LECTURE SQLITE ROBUSTE 🔥
# ================================

db_path = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-04/airflow/dags/e04_etl_pipeline_v01.db"

conn = sqlite3.connect(db_path)

# Vérification table
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
assert "multimodal_data" in tables["name"].values, "❌ Table absente dans SQLite"

df = pd.read_sql("SELECT * FROM multimodal_data", conn)
conn.close()

# ================================
# 🔍 DEBUG CRITIQUE
# ================================

print("\n🔍 DEBUG DATASET")
print("Shape:", df.shape)

assert len(df) > 9000, "❌ Dataset incomplet"
assert "kaggle_fake_news" in df["source"].values, "❌ Kaggle absent"

print("\nSources:")
print(df["source"].value_counts())

print("\nLabels:")
print(df["label_text"].value_counts())

# ================================
# 1️⃣ CONTEXTE DATASET
# ================================

total_rows = len(df)

dataset_type = "SCRAP ONLY"
if df["label_text"].isin(["fake", "real"]).sum() > 0:
    dataset_type = "MIXED (KAGGLE + SCRAP)"

print(f"\n🧠 Type dataset : {dataset_type}")
print(f"📈 Volume total : {total_rows}")

# ================================
# 2️⃣ QUALITÉ DATA
# ================================

text_ok = df["text"].fillna("").str.strip() != ""
text_quality_pct = text_ok.mean() * 100

published_ok = df["published"].notnull()
published_pct = published_ok.mean() * 100

url_ok = df["url"].fillna("").str.startswith("http")
url_quality_pct = url_ok.mean() * 100

# NOUVEAU KPI
duplicate_pct = df.duplicated(subset=["text"]).mean() * 100

# ================================
# 3️⃣ KPI MÉTIER
# ================================

image_pct = (df["nb_images"] > 0).mean() * 100
labeled_pct = df["label_text"].isin(["fake", "real"]).mean() * 100

label_distribution = df["label_text"].value_counts(normalize=True) * 100
source_distribution = df["source"].value_counts(normalize=True) * 100

# ================================
# 4️⃣ COMPLETUDE
# ================================

completeness = (df.notnull().mean() * 100).sort_values()

# ================================
# 5️⃣ PERFORMANCE
# ================================

execution_time = time.time() - start_time
rows_per_sec = total_rows / execution_time if execution_time > 0 else 0

# ================================
# 6️⃣ COÛT (SIMULATION 💰)
# ================================

cost_per_1000_rows = 0.05  # simulation cloud
estimated_cost = (total_rows / 1000) * cost_per_1000_rows

# ================================
# 7️⃣ SCORE GLOBAL
# ================================

data_quality_score = (
    text_quality_pct * 0.35 +
    published_pct * 0.15 +
    url_quality_pct * 0.2 +
    labeled_pct * 0.2 +
    (100 - duplicate_pct) * 0.1
)

# ================================
# 8️⃣ KPI STRUCTURÉS
# ================================

kpi = {
    "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "dataset_type": dataset_type,
    "total_rows": total_rows,
    "text_quality_pct": round(text_quality_pct, 2),
    "published_pct": round(published_pct, 2),
    "url_quality_pct": round(url_quality_pct, 2),
    "duplicate_pct": round(duplicate_pct, 2),
    "image_pct": round(image_pct, 2),
    "labeled_pct": round(labeled_pct, 2),
    "data_quality_score": round(data_quality_score, 2),
    "execution_time_sec": round(execution_time, 4),
    "rows_per_sec": round(rows_per_sec, 2),
    "estimated_cost_$": round(estimated_cost, 4)
}

print("\n📊 KPI globaux :")
for k, v in kpi.items():
    print(f"{k}: {v}")

# ================================
# 9️⃣ ALERTES INTELLIGENTES 🚨
# ================================

print("\n🚨 Alertes :")

if text_quality_pct < 70:
    print("⚠️ Texte faible → scraping à améliorer")

if labeled_pct < 70:
    print("⚠️ Dataset insuffisamment labellisé")

if published_pct < 50:
    print("⚠️ Dates manquantes")

if url_quality_pct < 80:
    print("⚠️ URLs invalides")

if duplicate_pct > 10:
    print("⚠️ Trop de doublons")

if data_quality_score < 65:
    print("🚨 Score global insuffisant")

# ================================
# 🔟 EXPORT KPI (MONITORING)
# ================================

# ================================
# LOAD DATASET FROM SQLITE 🔥
# ================================

BASE_PATH_E05 = "/content/drive/MyDrive/Colab Notebooks/Projet_12_Extrayez_des_données_multimodales_de_sites_web/Mission/Etape-05"
OUTPUT_PATH = f"{BASE_PATH_E05}/e05_kpi_monitoring.csv"

kpi_df = pd.DataFrame([kpi])
# kpi_df.to_csv(OUTPUT_PATH, mode="a", header=False, index=False)

file_exists = os.path.isfile(OUTPUT_PATH)

kpi_df.to_csv(
    OUTPUT_PATH,
    mode="a",
    header=not file_exists,
    index=False
)

print("\nKPI sauvegardés ici :", OUTPUT_PATH)

print()

# ================================
# 1️⃣1️⃣ DASHBOARD VISUEL
# ================================

# Complétude
plt.figure()
completeness.plot(kind="barh")
plt.title("Complétude des colonnes (%)")
plt.tight_layout()
plt.show()

print()

# Labels
plt.figure()
label_distribution.plot(kind="bar")
plt.title("Distribution des labels (%)")
plt.tight_layout()
plt.show()

print()

# Sources
plt.figure()
source_distribution.plot(kind="bar")
plt.title("Répartition des sources (%)")
plt.tight_layout()
plt.show()

print()

# KPI qualité
plt.figure()
pd.Series({
    "Text": text_quality_pct,
    "Published": published_pct,
    "URL": url_quality_pct,
    "Images": image_pct,
    "Labeled": labeled_pct,
    "No Duplicates": 100 - duplicate_pct
}).plot(kind="bar")
plt.title("KPI qualité (%)")
plt.tight_layout()
plt.show()

print()

# Score global
plt.figure()
pd.Series({"Data Quality Score": data_quality_score}).plot(kind="bar")
plt.title("Score global pipeline")
plt.tight_layout()
plt.show()

print()
````
